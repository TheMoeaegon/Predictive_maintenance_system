````python
import pandas as pd
import numpy as np
import zipfile
import io
from scipy import stats
from scipy.signal import welch
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
import warnings
warnings.filterwarnings('ignore')

class BearingRULPredictor:
    def __init__(self, zip_file_path):
        """
        Initialize the Bearing RUL Prediction system

        Args:
            zip_file_path: Path to the zip file containing CSV files
        """
        self.zip_file_path = zip_file_path
        self.raw_data = []
        self.features_df = None
        self.scaler = StandardScaler()
        self.model = None

        print("🔧 Bearing RUL Prediction System Initialized")

    def load_and_clean_data(self):
        """
        Load data from zip file and perform initial cleaning
        """
        print("\n📁 Loading and cleaning data from zip file...")

        with zipfile.ZipFile(self.zip_file_path, 'r') as zip_ref:
            file_list = sorted([f for f in zip_ref.namelist() if f.endswith('.csv')])
            print(f"Found {len(file_list)} CSV files")

            for idx, filename in enumerate(file_list):
                try:
                    # Read CSV from zip
                    with zip_ref.open(filename) as file:
                        df = pd.read_csv(file)

                    # Basic data cleaning
                    df = self._clean_single_file(df, idx)

                    # Store with metadata
                    self.raw_data.append({
                        'hour': idx,
                        'filename': filename,
                        'data': df,
                        'rul': 128 - idx  # RUL calculation
                    })

                    if idx % 20 == 0:
                        print(f"  Processed {idx+1}/{len(file_list)} files...")

                except Exception as e:
                    print(f"  ⚠️ Error processing {filename}: {e}")
                    continue

        print(f"✅ Successfully loaded {len(self.raw_data)} files")
        return len(self.raw_data)

    def _clean_single_file(self, df, hour):
        """
        Clean individual CSV file data

        Args:
            df: DataFrame from single CSV file
            hour: Hour index (0-128)
        """
        # Expected columns based on KAIST dataset description
        expected_columns = ['vibration_x', 'vibration_y', 'temperature_bearing', 'temperature_atmospheric']

        # Handle different possible column names
        if len(df.columns) == 4:
            df.columns = expected_columns

        # Remove any completely null rows
        df = df.dropna(how='all')

        # Handle outliers using IQR method
        for col in df.select_dtypes(include=[np.number]).columns:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

        # Forward fill any remaining NaN values
        df = df.fillna(method='ffill')
        df = df.fillna(method='bfill')

        return df

    def extract_features(self, window_size_seconds=10):
        """
        Extract features from raw signals using sliding windows

        Args:
            window_size_seconds: Size of window for feature extraction
        """
        print(f"\n🔍 Extracting features using {window_size_seconds}-second windows...")

        sampling_rate = 25600  # Hz
        samples_per_window = sampling_rate * window_size_seconds

        all_features = []

        for file_data in self.raw_data:
            hour = file_data['hour']
            df = file_data['data']
            rul = file_data['rul']

            # Calculate number of windows possible
            total_samples = len(df)
            num_windows = max(1, total_samples // samples_per_window)

            for window_idx in range(num_windows):
                start_idx = window_idx * samples_per_window
                end_idx = min(start_idx + samples_per_window, total_samples)

                window_data = df.iloc[start_idx:end_idx]

                if len(window_data) < samples_per_window * 0.8:  # Skip windows that are too small
                    continue

                # Extract features for this window
                features = self._extract_window_features(window_data, hour, window_idx, rul)
                all_features.append(features)

        self.features_df = pd.DataFrame(all_features)
        print(f"✅ Extracted features: {self.features_df.shape[0]} windows × {self.features_df.shape[1]-3} features")

        return self.features_df.shape

    def _extract_window_features(self, window_data, hour, window_idx, rul):
        """
        Extract comprehensive features from a single window
        """
        features = {
            'hour': hour,
            'window': window_idx,
            'rul': rul
        }

        # Process each sensor channel
        for col in ['vibration_x', 'vibration_y', 'temperature_bearing', 'temperature_atmospheric']:
            if col in window_data.columns:
                signal = window_data[col].values

                # Time domain features
                features.update(self._time_domain_features(signal, col))

                # Frequency domain features (only for vibration)
                if 'vibration' in col:
                    features.update(self._frequency_domain_features(signal, col))

        # Cross-signal features
        if 'vibration_x' in window_data.columns and 'vibration_y' in window_data.columns:
            features.update(self._cross_signal_features(
                window_data['vibration_x'].values,
                window_data['vibration_y'].values
            ))

        return features

    def _time_domain_features(self, signal, prefix):
        """Extract time-domain statistical features"""
        features = {}

        features[f'{prefix}_mean'] = np.mean(signal)
        features[f'{prefix}_std'] = np.std(signal)
        features[f'{prefix}_rms'] = np.sqrt(np.mean(signal**2))
        features[f'{prefix}_peak'] = np.max(np.abs(signal))
        features[f'{prefix}_peak_to_peak'] = np.ptp(signal)
        features[f'{prefix}_skewness'] = stats.skew(signal)
        features[f'{prefix}_kurtosis'] = stats.kurtosis(signal)
        features[f'{prefix}_variance'] = np.var(signal)

        # Crest factor
        rms_val = np.sqrt(np.mean(signal**2))
        features[f'{prefix}_crest_factor'] = np.max(np.abs(signal)) / rms_val if rms_val > 0 else 0

        # Shape factor
        mean_abs = np.mean(np.abs(signal))
        features[f'{prefix}_shape_factor'] = rms_val / mean_abs if mean_abs > 0 else 0

        return features

    def _frequency_domain_features(self, signal, prefix):
        """Extract frequency-domain features"""
        features = {}

        # Power Spectral Density
        freqs, psd = welch(signal, fs=25600, nperseg=1024)

        # Spectral features
        features[f'{prefix}_spectral_centroid'] = np.sum(freqs * psd) / np.sum(psd)
        features[f'{prefix}_spectral_rolloff'] = freqs[np.where(np.cumsum(psd) >= 0.85 * np.sum(psd))[0][0]]

        # Power in frequency bands (bearing fault frequencies)
        low_freq_power = np.sum(psd[(freqs >= 0) & (freqs <= 500)])
        mid_freq_power = np.sum(psd[(freqs > 500) & (freqs <= 2000)])
        high_freq_power = np.sum(psd[freqs > 2000])

        total_power = np.sum(psd)
        features[f'{prefix}_low_freq_power'] = low_freq_power / total_power
        features[f'{prefix}_mid_freq_power'] = mid_freq_power / total_power
        features[f'{prefix}_high_freq_power'] = high_freq_power / total_power

        return features

    def _cross_signal_features(self, vib_x, vib_y):
        """Extract features from relationships between signals"""
        features = {}

        # Correlation between vibration axes
        features['vib_xy_correlation'] = np.corrcoef(vib_x, vib_y)[0, 1]

        # Combined vibration magnitude
        combined_vib = np.sqrt(vib_x**2 + vib_y**2)
        features['combined_vib_rms'] = np.sqrt(np.mean(combined_vib**2))
        features['combined_vib_peak'] = np.max(combined_vib)

        return features

    def create_health_labels(self):
        """
        Create health classification labels based on RUL
        """
        if self.features_df is None:
            raise ValueError("Features not extracted yet. Run extract_features() first.")

        def classify_health(rul):
            if rul > 80:
                return 0  # Healthy
            elif rul > 40:
                return 1  # Degrading
            elif rul > 15:
                return 2  # Near Failure
            else:
                return 3  # Critical

        self.features_df['health_status'] = self.features_df['rul'].apply(classify_health)

        print("\n📊 Health Status Distribution:")
        print(self.features_df['health_status'].value_counts().sort_index())

        return self.features_df['health_status'].value_counts()

    def prepare_data(self, test_size=0.2, val_size=0.1):
        """
        Normalize features and split dataset
        """
        if self.features_df is None:
            raise ValueError("Features not extracted yet.")

        print(f"\n🔄 Preparing data with {test_size*100}% test, {val_size*100}% validation split...")

        # Separate features from labels
        feature_columns = [col for col in self.features_df.columns
                          if col not in ['hour', 'window', 'rul', 'health_status']]

        X = self.features_df[feature_columns]
        y_rul = self.features_df['rul']
        y_health = self.features_df['health_status']

        # Handle any remaining NaN values
        X = X.fillna(X.mean())

        # Split the data
        X_temp, self.X_test, y_rul_temp, self.y_rul_test, y_health_temp, self.y_health_test = \
            train_test_split(X, y_rul, y_health, test_size=test_size, random_state=42, stratify=y_health)

        val_size_adjusted = val_size / (1 - test_size)
        self.X_train, self.X_val, self.y_rul_train, self.y_rul_val, self.y_health_train, self.y_health_val = \
            train_test_split(X_temp, y_rul_temp, y_health_temp, test_size=val_size_adjusted,
                           random_state=42, stratify=y_health_temp)

        # Normalize features
        self.X_train_scaled = self.scaler.fit_transform(self.X_train)
        self.X_val_scaled = self.scaler.transform(self.X_val)
        self.X_test_scaled = self.scaler.transform(self.X_test)

        print(f"✅ Data prepared:")
        print(f"  Training: {self.X_train_scaled.shape[0]} samples")
        print(f"  Validation: {self.X_val_scaled.shape[0]} samples")
        print(f"  Testing: {self.X_test_scaled.shape[0]} samples")
        print(f"  Features: {self.X_train_scaled.shape[1]}")

        return {
            'train_shape': self.X_train_scaled.shape,
            'val_shape': self.X_val_scaled.shape,
            'test_shape': self.X_test_scaled.shape
        }

    def build_model(self, model_type='dense'):
        """
        Build neural network model for RUL prediction

        Args:
            model_type: 'dense' or 'lstm'
        """
        print(f"\n🏗️ Building {model_type.upper()} model...")

        if model_type == 'dense':
            self.model = Sequential([
                Dense(128, activation='relu', input_shape=(self.X_train_scaled.shape[1],)),
                Dropout(0.3),
                Dense(64, activation='relu'),
                Dropout(0.3),
                Dense(32, activation='relu'),
                Dropout(0.2),
                Dense(1, activation='linear')  # RUL prediction
            ])

        elif model_type == 'lstm':
            # Reshape data for LSTM (samples, timesteps, features)
            # We'll use the features as a sequence of length 1
            self.X_train_lstm = self.X_train_scaled.reshape((self.X_train_scaled.shape[0], 1, self.X_train_scaled.shape[1]))
            self.X_val_lstm = self.X_val_scaled.reshape((self.X_val_scaled.shape[0], 1, self.X_val_scaled.shape[1]))
            self.X_test_lstm = self.X_test_scaled.reshape((self.X_test_scaled.shape[0], 1, self.X_test_scaled.shape[1]))

            self.model = Sequential([
                LSTM(64, return_sequences=True, input_shape=(1, self.X_train_scaled.shape[1])),
                Dropout(0.3),
                LSTM(32, return_sequences=False),
                Dropout(0.3),
                Dense(16, activation='relu'),
                Dense(1, activation='linear')
            ])

        # Compile model
        self.model.compile(
            optimizer=Adam(learning_rate=0.001),
            loss='mse',
            metrics=['mae']
        )

        print("✅ Model built successfully")
        print(f"Total parameters: {self.model.count_params():,}")

        return self.model.summary()

    def train_model(self, epochs=100, batch_size=32, patience=15):
        """
        Train the model with early stopping
        """
        print(f"\n🚀 Training model for {epochs} epochs...")

        from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

        callbacks = [
            EarlyStopping(monitor='val_loss', patience=patience, restore_best_weights=True),
            ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=10, min_lr=0.0001)
        ]

        # Check if we're using LSTM
        if hasattr(self, 'X_train_lstm'):
            X_train, X_val = self.X_train_lstm, self.X_val_lstm
        else:
            X_train, X_val = self.X_train_scaled, self.X_val_scaled

        history = self.model.fit(
            X_train, self.y_rul_train,
            validation_data=(X_val, self.y_rul_val),
            epochs=epochs,
            batch_size=batch_size,
            callbacks=callbacks,
            verbose=1
        )

        print("✅ Training completed")
        return history

    def evaluate_model(self):
        """
        Evaluate the trained model
        """
        if self.model is None:
            raise ValueError("Model not trained yet.")

        print("\n📊 Evaluating model performance...")

        # Get test data
        if hasattr(self, 'X_test_lstm'):
            X_test = self.X_test_lstm
        else:
            X_test = self.X_test_scaled

        # Predictions
        y_pred_rul = self.model.predict(X_test, verbose=0).flatten()

        # RUL metrics
        mse = mean_squared_error(self.y_rul_test, y_pred_rul)
        mae = mean_absolute_error(self.y_rul_test, y_pred_rul)
        rmse = np.sqrt(mse)

        # Classification metrics (convert RUL to health status)
        def rul_to_health(rul):
            if rul > 80:
                return 0
            elif rul > 40:
                return 1
            elif rul > 15:
                return 2
            else:
                return 3

        y_pred_health = [rul_to_health(rul) for rul in y_pred_rul]
        y_true_health = [rul_to_health(rul) for rul in self.y_rul_test]

        accuracy = accuracy_score(y_true_health, y_pred_health)

        print(f"📈 RUL Prediction Metrics:")
        print(f"  MSE: {mse:.2f}")
        print(f"  MAE: {mae:.2f} hours")
        print(f"  RMSE: {rmse:.2f} hours")
        print(f"\n🎯 Health Classification Accuracy: {accuracy:.3f}")

        # Confusion Matrix
        cm = confusion_matrix(y_true_health, y_pred_health)

        plt.figure(figsize=(12, 5))

        # Plot 1: RUL Predictions vs Actual
        plt.subplot(1, 2, 1)
        plt.scatter(self.y_rul_test, y_pred_rul, alpha=0.6)
        plt.plot([self.y_rul_test.min(), self.y_rul_test.max()],
                [self.y_rul_test.min(), self.y_rul_test.max()], 'r--', lw=2)
        plt.xlabel('Actual RUL (hours)')
        plt.ylabel('Predicted RUL (hours)')
        plt.title(f'RUL Prediction Results\nMAE: {mae:.2f} hours')
        plt.grid(True, alpha=0.3)

        # Plot 2: Confusion Matrix
        plt.subplot(1, 2, 2)
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=['Healthy', 'Degrading', 'Near Failure', 'Critical'],
                   yticklabels=['Healthy', 'Degrading', 'Near Failure', 'Critical'])
        plt.title(f'Health Status Classification\nAccuracy: {accuracy:.3f}')
        plt.ylabel('True Status')
        plt.xlabel('Predicted Status')

        plt.tight_layout()
        plt.show()

        return {
            'mse': mse,
            'mae': mae,
            'rmse': rmse,
            'accuracy': accuracy,
            'confusion_matrix': cm
        }

    def predict_health(self, features):
        """
        Predict health status for new data

        Args:
            features: Feature vector or DataFrame
        """
        if self.model is None:
            raise ValueError("Model not trained yet.")

        # Ensure features are scaled
        if isinstance(features, pd.DataFrame):
            features_scaled = self.scaler.transform(features)
        else:
            features_scaled = self.scaler.transform(features.reshape(1, -1))

        # Reshape for LSTM if needed
        if hasattr(self, 'X_test_lstm'):
            features_scaled = features_scaled.reshape((features_scaled.shape[0], 1, features_scaled.shape[1]))

        # Predict RUL
        predicted_rul = self.model.predict(features_scaled, verbose=0).flatten()

        # Convert to health status
        def rul_to_health_status(rul):
            if rul > 80:
                return "HEALTHY"
            elif rul > 40:
                return "DEGRADING"
            elif rul > 15:
                return "NEAR_FAILURE"
            else:
                return "CRITICAL"

        status = [rul_to_health_status(rul) for rul in predicted_rul]

        return predicted_rul, status

def main():
    """
    Main execution function - modify this path to your zip file
    """
    # Initialize the system
    ZIP_FILE_PATH = "/content/drive/MyDrive/Vibration_Bearing_RuntoFailure.zip"


    predictor = BearingRULPredictor(ZIP_FILE_PATH)

    try:
        # Step 1: Load and clean data
        predictor.load_and_clean_data()

        # Step 2: Extract features
        predictor.extract_features(window_size_seconds=10)

        # Step 3: Create health labels
        predictor.create_health_labels()

        # Step 4: Prepare data
        predictor.prepare_data(test_size=0.2, val_size=0.1)

        # Step 5: Build model
        predictor.build_model(model_type='dense')  # Try 'lstm' for LSTM model

        # Step 6: Train model
        history = predictor.train_model(epochs=50, patience=10)

        # Step 7: Evaluate model
        results = predictor.evaluate_model()

        print("\n🎉 Training completed successfully!")
        print(f"Final MAE: {results['mae']:.2f} hours")
        print(f"Final Accuracy: {results['accuracy']:.3f}")

    except Exception as e:
        print(f"❌ Error: {e}")
        print("Please check your file paths and data format.")

if __name__ == "__main__":
    main()
````

````python
import pandas as pd
import numpy as np
import zipfile
import io
from scipy import stats
from scipy.signal import welch
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
import warnings
warnings.filterwarnings('ignore')

class BearingRULPredictor:
    def __init__(self, zip_file_path):
        """
        Initialize the Bearing RUL Prediction system

        Args:
            zip_file_path: Path to the zip file containing CSV files
        """
        self.zip_file_path = zip_file_path
        self.raw_data = []
        self.features_df = None
        self.scaler = StandardScaler()
        self.model = None

        print("🔧 Bearing RUL Prediction System Initialized")

    def load_and_clean_data(self):
        """
        Load data from zip file and perform initial cleaning
        """
        print("\n📁 Loading and cleaning data from zip file...")

        with zipfile.ZipFile(self.zip_file_path, 'r') as zip_ref:
            file_list = sorted([f for f in zip_ref.namelist() if f.endswith('.csv')])
            print(f"Found {len(file_list)} CSV files")

            for idx, filename in enumerate(file_list):
                try:
                    # Read CSV from zip
                    with zip_ref.open(filename) as file:
                        df = pd.read_csv(file)

                    # Basic data cleaning
                    df = self._clean_single_file(df, idx)

                    # Store with metadata
                    self.raw_data.append({
                        'hour': idx,
                        'filename': filename,
                        'data': df,
                        'rul': 128 - idx  # RUL calculation
                    })

                    if idx % 20 == 0:
                        print(f"  Processed {idx+1}/{len(file_list)} files...")

                except Exception as e:
                    print(f"  ⚠️ Error processing {filename}: {e}")
                    continue

        print(f"✅ Successfully loaded {len(self.raw_data)} files")
        return len(self.raw_data)

    def _clean_single_file(self, df, hour):
        """
        Clean individual CSV file data

        Args:
            df: DataFrame from single CSV file
            hour: Hour index (0-128)
        """
        # Expected columns based on KAIST dataset description
        expected_columns = ['vibration_x', 'vibration_y', 'temperature_bearing', 'temperature_atmospheric']

        # Handle different possible column names
        if len(df.columns) == 4:
            df.columns = expected_columns

        # Remove any completely null rows
        df = df.dropna(how='all')

        # Handle outliers using IQR method
        for col in df.select_dtypes(include=[np.number]).columns:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

        # Forward fill any remaining NaN values
        df = df.fillna(method='ffill')
        df = df.fillna(method='bfill')

        return df

    def extract_features(self, window_size_seconds=10, overlap_ratio=0.5):
        """
        Extract features from raw signals using overlapping sliding windows

        Args:
            window_size_seconds: Size of window for feature extraction
            overlap_ratio: Overlap ratio between consecutive windows (0-1)
        """
        print(f"\n🔍 Extracting features using {window_size_seconds}-second windows with {overlap_ratio*100}% overlap...")

        sampling_rate = 25600  # Hz
        samples_per_window = sampling_rate * window_size_seconds
        step_size = int(samples_per_window * (1 - overlap_ratio))  # Step size for overlapping

        all_features = []

        for file_data in self.raw_data:
            hour = file_data['hour']
            df = file_data['data']
            rul = file_data['rul']

            # Calculate overlapping windows
            total_samples = len(df)
            window_idx = 0

            for start_idx in range(0, total_samples - samples_per_window + 1, step_size):
                end_idx = start_idx + samples_per_window

                window_data = df.iloc[start_idx:end_idx]

                if len(window_data) < samples_per_window * 0.9:  # Skip windows that are too small
                    continue

                # Extract features for this window
                features = self._extract_window_features(window_data, hour, window_idx, rul)
                all_features.append(features)
                window_idx += 1

        self.features_df = pd.DataFrame(all_features)
        print(f"✅ Extracted features: {self.features_df.shape[0]} windows × {self.features_df.shape[1]-3} features")

        return self.features_df.shape

    def _extract_window_features(self, window_data, hour, window_idx, rul):
        """
        Extract comprehensive features from a single window
        """
        features = {
            'hour': hour,
            'window': window_idx,
            'rul': rul
        }

        # Process each sensor channel
        for col in ['vibration_x', 'vibration_y', 'temperature_bearing', 'temperature_atmospheric']:
            if col in window_data.columns:
                signal = window_data[col].values

                # Time domain features
                features.update(self._time_domain_features(signal, col))

                # Frequency domain features (only for vibration)
                if 'vibration' in col:
                    features.update(self._frequency_domain_features(signal, col))

        # Cross-signal features
        if 'vibration_x' in window_data.columns and 'vibration_y' in window_data.columns:
            features.update(self._cross_signal_features(
                window_data['vibration_x'].values,
                window_data['vibration_y'].values
            ))

        return features

    def _time_domain_features(self, signal, prefix):
        """Extract time-domain statistical features"""
        features = {}

        features[f'{prefix}_mean'] = np.mean(signal)
        features[f'{prefix}_std'] = np.std(signal)
        features[f'{prefix}_rms'] = np.sqrt(np.mean(signal**2))
        features[f'{prefix}_peak'] = np.max(np.abs(signal))
        features[f'{prefix}_peak_to_peak'] = np.ptp(signal)
        features[f'{prefix}_skewness'] = stats.skew(signal)
        features[f'{prefix}_kurtosis'] = stats.kurtosis(signal)
        features[f'{prefix}_variance'] = np.var(signal)

        # Crest factor
        rms_val = np.sqrt(np.mean(signal**2))
        features[f'{prefix}_crest_factor'] = np.max(np.abs(signal)) / rms_val if rms_val > 0 else 0

        # Shape factor
        mean_abs = np.mean(np.abs(signal))
        features[f'{prefix}_shape_factor'] = rms_val / mean_abs if mean_abs > 0 else 0

        return features

    def _frequency_domain_features(self, signal, prefix):
        """Extract frequency-domain features"""
        features = {}

        # Power Spectral Density
        freqs, psd = welch(signal, fs=25600, nperseg=1024)

        # Spectral features
        features[f'{prefix}_spectral_centroid'] = np.sum(freqs * psd) / np.sum(psd)
        features[f'{prefix}_spectral_rolloff'] = freqs[np.where(np.cumsum(psd) >= 0.85 * np.sum(psd))[0][0]]

        # Power in frequency bands (bearing fault frequencies)
        low_freq_power = np.sum(psd[(freqs >= 0) & (freqs <= 500)])
        mid_freq_power = np.sum(psd[(freqs > 500) & (freqs <= 2000)])
        high_freq_power = np.sum(psd[freqs > 2000])

        total_power = np.sum(psd)
        features[f'{prefix}_low_freq_power'] = low_freq_power / total_power
        features[f'{prefix}_mid_freq_power'] = mid_freq_power / total_power
        features[f'{prefix}_high_freq_power'] = high_freq_power / total_power

        return features

    def _cross_signal_features(self, vib_x, vib_y):
        """Extract features from relationships between signals"""
        features = {}

        # Correlation between vibration axes
        features['vib_xy_correlation'] = np.corrcoef(vib_x, vib_y)[0, 1]

        # Combined vibration magnitude
        combined_vib = np.sqrt(vib_x**2 + vib_y**2)
        features['combined_vib_rms'] = np.sqrt(np.mean(combined_vib**2))
        features['combined_vib_peak'] = np.max(combined_vib)

        return features

    def create_health_labels(self):
        """
        Create health classification labels and percentage-based RUL
        """
        if self.features_df is None:
            raise ValueError("Features not extracted yet. Run extract_features() first.")

        # Create percentage-based health indicator
        max_rul = self.features_df['rul'].max()  # 128 hours
        self.features_df['health_percentage'] = (self.features_df['rul'] / max_rul) * 100

        def classify_health(health_pct):
            if health_pct > 62.5:  # > 80 hours (62.5% of 128)
                return 0  # Healthy
            elif health_pct > 31.25:  # > 40 hours (31.25% of 128)
                return 1  # Degrading
            elif health_pct > 11.7:  # > 15 hours (11.7% of 128)
                return 2  # Near Failure
            else:
                return 3  # Critical

        self.features_df['health_status'] = self.features_df['health_percentage'].apply(classify_health)

        print("\n📊 Health Status Distribution:")
        print(self.features_df['health_status'].value_counts().sort_index())
        print(f"\n📊 Health Percentage Range: {self.features_df['health_percentage'].min():.1f}% - {self.features_df['health_percentage'].max():.1f}%")

        return self.features_df['health_status'].value_counts()

    def save_dataset(self, stage='features', save_path='/content/drive/MyDrive/BearingRUL_SavedData'):
        """
        Save dataset at different processing stages to Google Drive

        Args:
            stage: 'raw', 'features', 'processed', or 'split'
            save_path: Directory to save data (default: Google Drive)
        """
        import os
        import pickle
        from datetime import datetime
        import gc

        os.makedirs(save_path, exist_ok=True)
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

        if stage == 'raw':
            # Skip saving raw data to save RAM - it's too large
            print("⚠️ Skipping raw data save to prevent RAM crash - data is too large")
            print("✅ Raw data processing completed (not saved to preserve memory)")
            return None

        elif stage == 'features':
            # Save feature dataframe and free up raw data memory
            filename = f'features_dataset_{timestamp}.csv'
            filepath = os.path.join(save_path, filename)
            self.features_df.to_csv(filepath, index=False)
            print(f"✅ Features dataset saved to Google Drive: {filepath}")

            # Free up raw data memory after feature extraction
            if hasattr(self, 'raw_data') and self.raw_data:
                print("🗑️ Freeing raw data from memory to save RAM...")
                del self.raw_data
                self.raw_data = []
                gc.collect()
                print("✅ Memory freed")

        elif stage == 'processed':
            # Save features with labels
            filename = f'processed_dataset_{timestamp}.pkl'
            filepath = os.path.join(save_path, filename)
            processed_data = {
                'features_df': self.features_df,
                'target_type': getattr(self, 'target_type', 'rul'),
                'scaler': self.scaler if hasattr(self, 'scaler') else None
            }

            # Use protocol 4 for better memory efficiency
            with open(filepath, 'wb') as f:
                pickle.dump(processed_data, f, protocol=pickle.HIGHEST_PROTOCOL)
            print(f"✅ Processed dataset saved to Google Drive: {filepath}")

            # Clean up temporary variables
            del processed_data
            gc.collect()

        elif stage == 'split':
            # Save train/val/test splits
            filename = f'train_val_test_split_{timestamp}.pkl'
            filepath = os.path.join(save_path, filename)
            split_data = {
                'X_train': self.X_train_scaled,
                'X_val': self.X_val_scaled,
                'X_test': self.X_test_scaled,
                'y_train': self.y_target_train,
                'y_val': self.y_target_val,
                'y_test': self.y_target_test,
                'y_health_train': self.y_health_train,
                'y_health_val': self.y_health_val,
                'y_health_test': self.y_health_test,
                'scaler': self.scaler,
                'target_type': self.target_type,
                'feature_names': [col for col in self.features_df.columns
                                if col not in ['hour', 'window', 'rul', 'health_status', 'health_percentage']]
            }
            with open(filepath, 'wb') as f:
                pickle.dump(split_data, f)
            print(f"✅ Train/val/test splits saved to Google Drive: {filepath}")

        return filepath

    def load_dataset(self, filepath, stage='split'):
        """
        Load previously saved dataset

        Args:
            filepath: Path to saved data file
            stage: 'raw', 'features', 'processed', or 'split'
        """
        import pickle

        if stage == 'raw':
            with open(filepath, 'rb') as f:
                self.raw_data = pickle.load(f)
            print(f"✅ Raw data loaded from: {filepath}")

        elif stage == 'features':
            self.features_df = pd.read_csv(filepath)
            print(f"✅ Features loaded from: {filepath}")

        elif stage == 'processed':
            with open(filepath, 'rb') as f:
                data = pickle.load(f)
            self.features_df = data['features_df']
            self.target_type = data.get('target_type', 'rul')
            if data['scaler']:
                self.scaler = data['scaler']
            print(f"✅ Processed data loaded from: {filepath}")

        elif stage == 'split':
            with open(filepath, 'rb') as f:
                data = pickle.load(f)

            self.X_train_scaled = data['X_train']
            self.X_val_scaled = data['X_val']
            self.X_test_scaled = data['X_test']
            self.y_target_train = data['y_train']
            self.y_target_val = data['y_val']
            self.y_target_test = data['y_test']
            self.y_health_train = data['y_health_train']
            self.y_health_val = data['y_health_val']
            self.y_health_test = data['y_health_test']
            self.scaler = data['scaler']
            self.target_type = data['target_type']

            print(f"✅ Train/val/test splits loaded from: {filepath}")
            print(f"  Training: {self.X_train_scaled.shape[0]} samples")
            print(f"  Validation: {self.X_val_scaled.shape[0]} samples")
            print(f"  Testing: {self.X_test_scaled.shape[0]} samples")

        return True

    def prepare_data(self, test_size=0.2, val_size=0.1, use_percentage=True):
        """
        Normalize features and split dataset

        Args:
            test_size: Proportion for test set
            val_size: Proportion for validation set
            use_percentage: Whether to use health_percentage instead of raw RUL
        """
        if self.features_df is None:
            raise ValueError("Features not extracted yet.")

        target_col = 'health_percentage' if use_percentage else 'rul'
        print(f"\n🔄 Preparing data with {test_size*100}% test, {val_size*100}% validation split...")
        print(f"Using target: {target_col}")

        # Separate features from labels
        feature_columns = [col for col in self.features_df.columns
                          if col not in ['hour', 'window', 'rul', 'health_status', 'health_percentage']]

        X = self.features_df[feature_columns]
        y_target = self.features_df[target_col]
        y_health = self.features_df['health_status']

        # Handle any remaining NaN values
        X = X.fillna(X.mean())

        # Split the data
        X_temp, self.X_test, y_target_temp, self.y_target_test, y_health_temp, self.y_health_test = \
            train_test_split(X, y_target, y_health, test_size=test_size, random_state=42, stratify=y_health)

        val_size_adjusted = val_size / (1 - test_size)
        self.X_train, self.X_val, self.y_target_train, self.y_target_val, self.y_health_train, self.y_health_val = \
            train_test_split(X_temp, y_target_temp, y_health_temp, test_size=val_size_adjusted,
                           random_state=42, stratify=y_health_temp)

        # Store which target we're using
        self.target_type = target_col

        # Normalize features
        self.X_train_scaled = self.scaler.fit_transform(self.X_train)
        self.X_val_scaled = self.scaler.transform(self.X_val)
        self.X_test_scaled = self.scaler.transform(self.X_test)

        print(f"✅ Data prepared:")
        print(f"  Training: {self.X_train_scaled.shape[0]} samples")
        print(f"  Validation: {self.X_val_scaled.shape[0]} samples")
        print(f"  Testing: {self.X_test_scaled.shape[0]} samples")
        print(f"  Features: {self.X_train_scaled.shape[1]}")

        return {
            'train_shape': self.X_train_scaled.shape,
            'val_shape': self.X_val_scaled.shape,
            'test_shape': self.X_test_scaled.shape
        }

    def build_model(self, model_type='dense'):
        """
        Build improved neural network model for RUL prediction

        Args:
            model_type: 'dense' or 'lstm'
        """
        print(f"\n🏗️ Building improved {model_type.upper()} model...")

        if model_type == 'dense':
            self.model = Sequential([
                Dense(256, activation='relu', input_shape=(self.X_train_scaled.shape[1],)),
                Dropout(0.4),
                Dense(128, activation='relu'),
                Dropout(0.3),
                Dense(64, activation='relu'),
                Dropout(0.2),
                Dense(32, activation='relu'),
                Dropout(0.1),
                Dense(1, activation='linear')  # RUL/percentage prediction
            ])

        elif model_type == 'lstm':
            # Reshape data for LSTM (samples, timesteps, features)
            self.X_train_lstm = self.X_train_scaled.reshape((self.X_train_scaled.shape[0], 1, self.X_train_scaled.shape[1]))
            self.X_val_lstm = self.X_val_scaled.reshape((self.X_val_scaled.shape[0], 1, self.X_val_scaled.shape[1]))
            self.X_test_lstm = self.X_test_scaled.reshape((self.X_test_scaled.shape[0], 1, self.X_test_scaled.shape[1]))

            self.model = Sequential([
                LSTM(128, return_sequences=True, input_shape=(1, self.X_train_scaled.shape[1])),
                Dropout(0.4),
                LSTM(64, return_sequences=False),
                Dropout(0.3),
                Dense(32, activation='relu'),
                Dropout(0.2),
                Dense(1, activation='linear')
            ])

        # Compile model with improved settings
        self.model.compile(
            optimizer=Adam(learning_rate=0.0005),  # Slightly lower learning rate
            loss='mse',
            metrics=['mae']
        )

        print("✅ Model built successfully")
        print(f"Total parameters: {self.model.count_params():,}")

        return self.model.summary()

    def train_model(self, epochs=100, batch_size=32, patience=20, save_best=True):
        """
        Train the model with improved early stopping and learning rate scheduling
        """
        print(f"\n🚀 Training model for {epochs} epochs...")

        from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
        import os

        # Create models directory in Google Drive if it doesn't exist
        models_dir = '/content/drive/MyDrive/BearingRUL_Models'
        os.makedirs(models_dir, exist_ok=True)

        callbacks = [
            EarlyStopping(monitor='val_loss', patience=patience, restore_best_weights=True, verbose=1),
            ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=8, min_lr=0.00001, verbose=1)
        ]

        # Add model checkpoint to save best model to Google Drive
        if save_best:
            model_path = os.path.join(models_dir, f'best_model_{self.target_type}.h5')
            callbacks.append(
                ModelCheckpoint(
                    filepath=model_path,
                    monitor='val_loss',
                    save_best_only=True,
                    save_weights_only=False,
                    verbose=1
                )
            )
            print(f"Best model will be saved to Google Drive: {model_path}")

        # Check if we're using LSTM
        if hasattr(self, 'X_train_lstm'):
            X_train, X_val = self.X_train_lstm, self.X_val_lstm
        else:
            X_train, X_val = self.X_train_scaled, self.X_val_scaled

        history = self.model.fit(
            X_train, self.y_target_train,  # Use new target variable
            validation_data=(X_val, self.y_target_val),
            epochs=epochs,
            batch_size=batch_size,
            callbacks=callbacks,
            verbose=1
        )

        print("✅ Training completed")
        return history

    def evaluate_model(self):
        """
        Evaluate the trained model - FIXED BUG
        """
        if self.model is None:
            raise ValueError("Model not trained yet.")

        print("\n📊 Evaluating model performance...")

        # Get test data
        if hasattr(self, 'X_test_lstm'):
            X_test = self.X_test_lstm
        else:
            X_test = self.X_test_scaled

        # Predictions
        y_pred_target = self.model.predict(X_test, verbose=0).flatten()

        # RUL metrics - FIXED: Use correct target variable
        mse = mean_squared_error(self.y_target_test, y_pred_target)
        mae = mean_absolute_error(self.y_target_test, y_pred_target)
        rmse = np.sqrt(mse)

        # Convert predictions to RUL for health classification
        if self.target_type == 'health_percentage':
            # Convert percentage back to RUL hours
            max_rul = 128  # Original max RUL
            y_pred_rul = (y_pred_target / 100) * max_rul
            y_true_rul = (self.y_target_test / 100) * max_rul
        else:
            # Already in RUL format
            y_pred_rul = y_pred_target
            y_true_rul = self.y_target_test

        # Classification metrics (convert RUL to health status)
        def rul_to_health(rul):
            if rul > 80:
                return 0  # Healthy
            elif rul > 40:
                return 1  # Degrading
            elif rul > 15:
                return 2  # Near Failure
            else:
                return 3  # Critical

        y_pred_health = [rul_to_health(rul) for rul in y_pred_rul]
        y_true_health = [rul_to_health(rul) for rul in y_true_rul]

        accuracy = accuracy_score(y_true_health, y_pred_health)

        # Display target-specific metrics
        if self.target_type == 'health_percentage':
            print(f"📈 Health Percentage Prediction Metrics:")
            print(f"  MSE: {mse:.2f}")
            print(f"  MAE: {mae:.2f}%")
            print(f"  RMSE: {rmse:.2f}%")
        else:
            print(f"📈 RUL Prediction Metrics:")
            print(f"  MSE: {mse:.2f}")
            print(f"  MAE: {mae:.2f} hours")
            print(f"  RMSE: {rmse:.2f} hours")

        print(f"\n🎯 Health Classification Accuracy: {accuracy:.3f}")

        # Confusion Matrix
        cm = confusion_matrix(y_true_health, y_pred_health)

        plt.figure(figsize=(12, 5))

        # Plot 1: Target Predictions vs Actual
        plt.subplot(1, 2, 1)
        plt.scatter(self.y_target_test, y_pred_target, alpha=0.6)
        plt.plot([self.y_target_test.min(), self.y_target_test.max()],
                [self.y_target_test.min(), self.y_target_test.max()], 'r--', lw=2)

        if self.target_type == 'health_percentage':
            plt.xlabel('Actual Health Percentage (%)')
            plt.ylabel('Predicted Health Percentage (%)')
            plt.title(f'Health Percentage Prediction Results\nMAE: {mae:.2f}%')
        else:
            plt.xlabel('Actual RUL (hours)')
            plt.ylabel('Predicted RUL (hours)')
            plt.title(f'RUL Prediction Results\nMAE: {mae:.2f} hours')

        plt.grid(True, alpha=0.3)

        # Plot 2: Confusion Matrix
        plt.subplot(1, 2, 2)
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=['Healthy', 'Degrading', 'Near Failure', 'Critical'],
                   yticklabels=['Healthy', 'Degrading', 'Near Failure', 'Critical'])
        plt.title(f'Health Status Classification\nAccuracy: {accuracy:.3f}')
        plt.ylabel('True Status')
        plt.xlabel('Predicted Status')

        plt.tight_layout()
        plt.show()

        return {
            'mse': mse,
            'mae': mae,
            'rmse': rmse,
            'accuracy': accuracy,
            'confusion_matrix': cm
        }

    def predict_health(self, features):
        """
        Predict health status for new data

        Args:
            features: Feature vector or DataFrame
        """
        if self.model is None:
            raise ValueError("Model not trained yet.")

        # Ensure features are scaled
        if isinstance(features, pd.DataFrame):
            features_scaled = self.scaler.transform(features)
        else:
            features_scaled = self.scaler.transform(features.reshape(1, -1))

        # Reshape for LSTM if needed
        if hasattr(self, 'X_test_lstm'):
            features_scaled = features_scaled.reshape((features_scaled.shape[0], 1, features_scaled.shape[1]))

        # Predict target
        predicted_target = self.model.predict(features_scaled, verbose=0).flatten()

        # Convert to RUL if needed
        if self.target_type == 'health_percentage':
            predicted_rul = (predicted_target / 100) * 128  # Convert percentage to RUL
        else:
            predicted_rul = predicted_target

        # Convert to health status
        def rul_to_health_status(rul):
            if rul > 80:
                return "HEALTHY"
            elif rul > 40:
                return "DEGRADING"
            elif rul > 15:
                return "NEAR_FAILURE"
            else:
                return "CRITICAL"

        status = [rul_to_health_status(rul) for rul in predicted_rul]

        return predicted_rul, status

    def save_model_components(self, save_path='/content/drive/MyDrive/BearingRUL_Models'):
        """
        Save complete model with all components for deployment to Google Drive
        """
        import os
        import pickle
        from datetime import datetime
        from tensorflow.keras.models import save_model

        os.makedirs(save_path, exist_ok=True)
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

        # Save trained model
        model_path = os.path.join(save_path, f'trained_model_{timestamp}.h5')
        self.model.save(model_path)

        # Save scaler and metadata
        components_path = os.path.join(save_path, f'model_components_{timestamp}.pkl')
        components = {
            'scaler': self.scaler,
            'target_type': self.target_type,
            'feature_names': [col for col in self.features_df.columns
                            if col not in ['hour', 'window', 'rul', 'health_status', 'health_percentage']],
            'health_thresholds': {
                'healthy': 62.5,
                'degrading': 31.25,
                'near_failure': 11.7
            },
            'model_architecture': 'dense' if not hasattr(self, 'X_train_lstm') else 'lstm'
        }

        with open(components_path, 'wb') as f:
            pickle.dump(components, f)

        print(f"Model saved to Google Drive: {model_path}")
        print(f"Components saved to Google Drive: {components_path}")

        return model_path, components_path

    def load_trained_model(self, model_path, components_path):
        """
        Load complete trained model for deployment
        """
        import pickle
        from tensorflow.keras.models import load_model

        # Load model
        self.model = load_model(model_path)

        # Load components
        with open(components_path, 'rb') as f:
            components = pickle.load(f)

        self.scaler = components['scaler']
        self.target_type = components['target_type']
        self.feature_names = components['feature_names']
        self.health_thresholds = components['health_thresholds']

        print(f"Model loaded from: {model_path}")
        print(f"Components loaded from: {components_path}")
        print(f"Target type: {self.target_type}")
        print(f"Features: {len(self.feature_names)}")

        return True

def main():
    """
    Main execution function - modify this path to your zip file
    """
    # Initialize the system
    ZIP_FILE_PATH = "/content/drive/MyDrive/Vibration_Bearing_RuntoFailure.zip"  # Update this path

    predictor = BearingRULPredictor(ZIP_FILE_PATH)

    try:
        # Step 1: Load and clean data
        predictor.load_and_clean_data()

        # Save raw data to Google Drive after cleaning
        predictor.save_dataset(stage='raw')

        # Step 2: Extract features
        predictor.extract_features(window_size_seconds=10)

        # Save features to Google Drive
        predictor.save_dataset(stage='features')

        # Step 3: Create health labels
        predictor.create_health_labels()

        # Save processed data to Google Drive
        predictor.save_dataset(stage='processed')

        # Step 4: Prepare data
        predictor.prepare_data(test_size=0.2, val_size=0.1)

        # Save train/val/test splits to Google Drive
        predictor.save_dataset(stage='split')

        # Step 5: Build model
        predictor.build_model(model_type='dense')  # Try 'lstm' for LSTM model

        # Step 6: Train model (will save best model to Google Drive automatically)
        history = predictor.train_model(epochs=50, patience=10)

        # Step 7: Evaluate model (FIXED)
        results = predictor.evaluate_model()

        # Step 8: Save final model components to Google Drive
        model_path, components_path = predictor.save_model_components()

        print("\n🎉 Training completed successfully!")
        print(f"Final MAE: {results['mae']:.2f}")
        print(f"Final Accuracy: {results['accuracy']:.3f}")
        print(f"\n💾 All files saved to Google Drive:")
        print(f"  - Best model: {model_path}")
        print(f"  - Model components: {components_path}")

    except Exception as e:
        print(f"❌ Error: {e}")
        print("Please check your file paths and data format.")

if __name__ == "__main__":
    main()
````

````python
import pandas as pd
import numpy as np
import zipfile
import io
from scipy import stats
from scipy.signal import welch
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
import warnings
warnings.filterwarnings('ignore')

class BearingRULPredictor:
    def __init__(self, zip_file_path=None):
        """
        Initialize the Bearing RUL Prediction system

        Args:
            zip_file_path: Path to the zip file containing CSV files (optional if loading preprocessed data)
        """
        self.zip_file_path = zip_file_path
        self.raw_data = []
        self.features_df = None
        self.scaler = StandardScaler()
        self.model = None

        print("🔧 Bearing RUL Prediction System Initialized")

    def load_and_clean_data(self):
        """
        Load data from zip file and perform initial cleaning
        """
        print("\n📁 Loading and cleaning data from zip file...")

        with zipfile.Zipfile(self.zip_file_path, 'r') as zip_ref:
            file_list = sorted([f for f in zip_ref.namelist() if f.endswith('.csv')])
            print(f"Found {len(file_list)} CSV files")

            for idx, filename in enumerate(file_list):
                try:
                    # Read CSV from zip
                    with zip_ref.open(filename) as file:
                        df = pd.read_csv(file)

                    # Basic data cleaning
                    df = self._clean_single_file(df, idx)

                    # Store with metadata
                    self.raw_data.append({
                        'hour': idx,
                        'filename': filename,
                        'data': df,
                        'rul': 128 - idx  # RUL calculation
                    })

                    if idx % 20 == 0:
                        print(f"  Processed {idx+1}/{len(file_list)} files...")

                except Exception as e:
                    print(f"  ⚠️ Error processing {filename}: {e}")
                    continue

        print(f"✅ Successfully loaded {len(self.raw_data)} files")
        return len(self.raw_data)

    def _clean_single_file(self, df, hour):
        """
        Clean individual CSV file data

        Args:
            df: DataFrame from single CSV file
            hour: Hour index (0-128)
        """
        # Expected columns based on KAIST dataset description
        expected_columns = ['vibration_x', 'vibration_y', 'temperature_bearing', 'temperature_atmospheric']

        # Handle different possible column names
        if len(df.columns) == 4:
            df.columns = expected_columns

        # Remove any completely null rows
        df = df.dropna(how='all')

        # Handle outliers using IQR method
        for col in df.select_dtypes(include=[np.number]).columns:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

        # Forward fill any remaining NaN values
        df = df.fillna(method='ffill')
        df = df.fillna(method='bfill')

        return df

    def extract_features(self, window_size_seconds=10, overlap_ratio=0.5):
        """
        Extract features from raw signals using overlapping sliding windows

        Args:
            window_size_seconds: Size of window for feature extraction
            overlap_ratio: Overlap ratio between consecutive windows (0-1)
        """
        print(f"\n🔍 Extracting features using {window_size_seconds}-second windows with {overlap_ratio*100}% overlap...")

        sampling_rate = 25600  # Hz
        samples_per_window = sampling_rate * window_size_seconds
        step_size = int(samples_per_window * (1 - overlap_ratio))  # Step size for overlapping

        all_features = []

        for file_data in self.raw_data:
            hour = file_data['hour']
            df = file_data['data']
            rul = file_data['rul']

            # Calculate overlapping windows
            total_samples = len(df)
            window_idx = 0

            for start_idx in range(0, total_samples - samples_per_window + 1, step_size):
                end_idx = start_idx + samples_per_window

                window_data = df.iloc[start_idx:end_idx]

                if len(window_data) < samples_per_window * 0.9:  # Skip windows that are too small
                    continue

                # Extract features for this window
                features = self._extract_window_features(window_data, hour, window_idx, rul)
                all_features.append(features)
                window_idx += 1

        self.features_df = pd.DataFrame(all_features)
        print(f"✅ Extracted features: {self.features_df.shape[0]} windows × {self.features_df.shape[1]-3} features")

        return self.features_df.shape

    def _extract_window_features(self, window_data, hour, window_idx, rul):
        """
        Extract comprehensive features from a single window
        """
        features = {
            'hour': hour,
            'window': window_idx,
            'rul': rul
        }

        # Process each sensor channel
        for col in ['vibration_x', 'vibration_y', 'temperature_bearing', 'temperature_atmospheric']:
            if col in window_data.columns:
                signal = window_data[col].values

                # Time domain features
                features.update(self._time_domain_features(signal, col))

                # Frequency domain features (only for vibration)
                if 'vibration' in col:
                    features.update(self._frequency_domain_features(signal, col))

        # Cross-signal features
        if 'vibration_x' in window_data.columns and 'vibration_y' in window_data.columns:
            features.update(self._cross_signal_features(
                window_data['vibration_x'].values,
                window_data['vibration_y'].values
            ))

        return features

    def create_health_labels(self):
        """
        Create health classification labels and percentage-based RUL
        """
        if self.features_df is None:
            raise ValueError("Features not extracted yet. Run extract_features() first.")

        # Create percentage-based health indicator
        max_rul = self.features_df['rul'].max()  # 128 hours
        self.features_df['health_percentage'] = (self.features_df['rul'] / max_rul) * 100

        def classify_health(health_pct):
            if health_pct > 62.5:  # > 80 hours (62.5% of 128)
                return 0  # Healthy
            elif health_pct > 31.25:  # > 40 hours (31.25% of 128)
                return 1  # Degrading
            elif health_pct > 11.7:  # > 15 hours (11.7% of 128)
                return 2  # Near Failure
            else:
                return 3  # Critical

        self.features_df['health_status'] = self.features_df['health_percentage'].apply(classify_health)

        print("\n📊 Health Status Distribution:")
        print(self.features_df['health_status'].value_counts().sort_index())
        print(f"\n📊 Health Percentage Range: {self.features_df['health_percentage'].min():.1f}% - {self.features_df['health_percentage'].max():.1f}%")

        return self.features_df['health_status'].value_counts()

    def save_dataset(self, stage='features', save_path='/content/drive/MyDrive/BearingRUL_SavedData'):
        """
        Save dataset at different processing stages to Google Drive

        Args:
            stage: 'raw', 'features', 'processed', or 'split'
            save_path: Directory to save data (default: Google Drive)
        """
        import os
        import pickle
        from datetime import datetime
        import gc

        os.makedirs(save_path, exist_ok=True)
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

        if stage == 'raw':
            # Skip saving raw data to save RAM - it's too large
            print("⚠️ Skipping raw data save to prevent RAM crash - data is too large")
            print("✅ Raw data processing completed (not saved to preserve memory)")
            return None

        elif stage == 'features':
            # Save feature dataframe and free up raw data memory
            filename = f'features_dataset_{timestamp}.csv'
            filepath = os.path.join(save_path, filename)
            self.features_df.to_csv(filepath, index=False)
            print(f"✅ Features dataset saved to Google Drive: {filepath}")

            # Free up raw data memory after feature extraction
            if hasattr(self, 'raw_data') and self.raw_data:
                print("🗑️ Freeing raw data from memory to save RAM...")
                del self.raw_data
                self.raw_data = []
                gc.collect()
                print("✅ Memory freed")

        elif stage == 'processed':
            # Save features with labels
            filename = f'processed_dataset_{timestamp}.pkl'
            filepath = os.path.join(save_path, filename)
            processed_data = {
                'features_df': self.features_df,
                'target_type': getattr(self, 'target_type', 'rul'),
                'scaler': self.scaler if hasattr(self, 'scaler') else None
            }

            # Use protocol 4 for better memory efficiency
            with open(filepath, 'wb') as f:
                pickle.dump(processed_data, f, protocol=pickle.HIGHEST_PROTOCOL)
            print(f"✅ Processed dataset saved to Google Drive: {filepath}")

            # Clean up temporary variables
            del processed_data
            gc.collect()

        elif stage == 'split':
            # Save train/val/test splits
            filename = f'train_val_test_split_{timestamp}.pkl'
            filepath = os.path.join(save_path, filename)
            split_data = {
                'X_train': self.X_train_scaled,
                'X_val': self.X_val_scaled,
                'X_test': self.X_test_scaled,
                'y_train': self.y_target_train,
                'y_val': self.y_target_val,
                'y_test': self.y_target_test,
                'y_health_train': self.y_health_train,
                'y_health_val': self.y_health_val,
                'y_health_test': self.y_health_test,
                'scaler': self.scaler,
                'target_type': self.target_type,
                # Include feature_names for reconstruction
                'feature_names': data.get('feature_names', []) # Get feature names from loaded data if available
            }
            # Add feature names if not present in loaded data (for backward compatibility)
            if not split_data['feature_names'] and hasattr(self, 'features_df') and self.features_df is not None:
                 split_data['feature_names'] = [col for col in self.features_df.columns
                                                if col not in ['hour', 'window', 'rul', 'health_status', 'health_percentage']]


            with open(filepath, 'wb') as f:
                pickle.dump(split_data, f)
            print(f"✅ Train/val/test splits saved to Google Drive: {filepath}")

        return filepath

    def load_dataset(self, filepath, stage='split'):
        """
        Load previously saved dataset

        Args:
            filepath: Path to saved data file
            stage: 'raw', 'features', 'processed', or 'split'
        """
        import pickle

        if stage == 'raw':
            with open(filepath, 'rb') as f:
                self.raw_data = pickle.load(f)
            print(f"✅ Raw data loaded from: {filepath}")

        elif stage == 'features':
            self.features_df = pd.read_csv(filepath)
            print(f"✅ Features loaded from: {filepath}")

        elif stage == 'processed':
            with open(filepath, 'rb') as f:
                data = pickle.load(f)
            self.features_df = data['features_df']
            self.target_type = data.get('target_type', 'rul')
            if data['scaler']:
                self.scaler = data['scaler']
            print(f"✅ Processed data loaded from: {filepath}")

        elif stage == 'split':
            with open(filepath, 'rb') as f:
                data = pickle.load(f)

            self.X_train_scaled = data['X_train']
            self.X_val_scaled = data['X_val']
            self.X_test_scaled = data['X_test']
            self.y_target_train = data['y_train']
            self.y_target_val = data['y_val']
            self.y_target_test = data['y_test']
            self.y_health_train = data['y_health_train']
            self.y_health_val = data['y_health_val']
            self.y_health_test = data['y_health_test']
            self.scaler = data['scaler']
            self.target_type = data['target_type']

            # Reconstruct features_df from split data
            feature_names = data['feature_names']
            # Concatenate scaled data
            all_scaled_X = np.vstack([self.X_train_scaled, self.X_val_scaled, self.X_test_scaled])
            self.features_df = pd.DataFrame(all_scaled_X, columns=feature_names)

            # Add back metadata columns (approximate reconstruction)
            all_y_target = np.concatenate([self.y_target_train, self.y_target_val, self.y_target_test])
            all_y_health = np.concatenate([self.y_health_train, self.y_health_val, self.y_health_test])

            if self.target_type == 'health_percentage':
                self.features_df['health_percentage'] = all_y_target
                # Approximate RUL from health percentage
                self.features_df['rul'] = (all_y_target / 100) * 128
            else: # target_type is 'rul'
                self.features_df['rul'] = all_y_target
                # Approximate health percentage from RUL
                max_rul = 128 # Assuming original max RUL was 128
                self.features_df['health_percentage'] = (all_y_target / max_rul) * 100

            self.features_df['health_status'] = all_y_health
            # Add hour and window - these cannot be perfectly reconstructed from split data alone,
            # so we'll add placeholder values or attempt a simple approximation if the original
            # structure (e.g., sorted by hour/window) is preserved in the split.
            # For now, adding simple index as placeholder. A more robust approach would save
            # hour/window in the split data if needed for saving features_df.
            # Let's add placeholder columns to avoid errors in save_model_components
            if 'hour' not in self.features_df.columns:
                 self.features_df['hour'] = -1 # Placeholder
            if 'window' not in self.features_df.columns:
                 self.features_df['window'] = range(len(self.features_df)) # Placeholder


            print(f"✅ Train/val/test splits loaded from: {filepath}")
            print(f"  Training: {self.X_train_scaled.shape[0]} samples")
            print(f"  Validation: {self.X_val_scaled.shape[0]} samples")
            print(f"  Testing: {self.X_test_scaled.shape[0]} samples")

        return True

    def prepare_data(self, test_size=0.2, val_size=0.1, use_percentage=True):
        """
        Normalize features and split dataset

        Args:
            test_size: Proportion for test set
            val_size: Proportion for validation set
            use_percentage: Whether to use health_percentage instead of raw RUL
        """
        if self.features_df is None:
            raise ValueError("Features not extracted yet.")

        target_col = 'health_percentage' if use_percentage else 'rul'
        print(f"\n🔄 Preparing data with {test_size*100}% test, {val_size*100}% validation split...")
        print(f"Using target: {target_col}")

        # Separate features from labels
        feature_columns = [col for col in self.features_df.columns
                          if col not in ['hour', 'window', 'rul', 'health_status', 'health_percentage']]

        X = self.features_df[feature_columns]
        y_target = self.features_df[target_col]
        y_health = self.features_df['health_status']

        # Handle any remaining NaN values
        X = X.fillna(X.mean())

        # Split the data
        X_temp, self.X_test, y_target_temp, self.y_target_test, y_health_temp, self.y_health_test = \
            train_test_split(X, y_target, y_health, test_size=test_size, random_state=42, stratify=y_health)

        val_size_adjusted = val_size / (1 - test_size)
        self.X_train, self.X_val, self.y_target_train, self.y_target_val, self.y_health_train, self.y_health_val = \
            train_test_split(X_temp, y_target_temp, y_health_temp, test_size=val_size_adjusted,
                           random_state=42, stratify=y_health_temp)

        # Store which target we're using
        self.target_type = target_col

        # Normalize features
        self.X_train_scaled = self.scaler.fit_transform(self.X_train)
        self.X_val_scaled = self.scaler.transform(self.X_val)
        self.X_test_scaled = self.scaler.transform(self.X_test)

        print(f"✅ Data prepared:")
        print(f"  Training: {self.X_train_scaled.shape[0]} samples")
        print(f"  Validation: {self.X_val_scaled.shape[0]} samples")
        print(f"  Testing: {self.X_test_scaled.shape[0]} samples")
        print(f"  Features: {self.X_train_scaled.shape[1]}")

        return {
            'train_shape': self.X_train_scaled.shape,
            'val_shape': self.X_val_scaled.shape,
            'test_shape': self.X_test_scaled.shape
        }

    def build_model(self, model_type='dense'):
        """
        Build improved neural network model for RUL prediction

        Args:
            model_type: 'dense' or 'lstm'
        """
        print(f"\n🏗️ Building improved {model_type.upper()} model...")

        if model_type == 'dense':
            self.model = Sequential([
                Dense(256, activation='relu', input_shape=(self.X_train_scaled.shape[1],)),
                Dropout(0.4),
                Dense(128, activation='relu'),
                Dropout(0.3),
                Dense(64, activation='relu'),
                Dropout(0.2),
                Dense(32, activation='relu'),
                Dropout(0.1),
                Dense(1, activation='linear')  # RUL/percentage prediction
            ])

        elif model_type == 'lstm':
            # Reshape data for LSTM (samples, timesteps, features)
            self.X_train_lstm = self.X_train_scaled.reshape((self.X_train_scaled.shape[0], 1, self.X_train_scaled.shape[1]))
            self.X_val_lstm = self.X_val_scaled.reshape((self.X_val_scaled.shape[0], 1, self.X_val_scaled.shape[1]))
            self.X_test_lstm = self.X_test_scaled.reshape((self.X_test_scaled.shape[0], 1, self.X_test_scaled.shape[1]))

            self.model = Sequential([
                LSTM(128, return_sequences=True, input_shape=(1, self.X_train_scaled.shape[1])),
                Dropout(0.4),
                LSTM(64, return_sequences=False),
                Dropout(0.3),
                Dense(32, activation='relu'),
                Dropout(0.2),
                Dense(1, activation='linear')
            ])

        # Compile model with improved settings
        self.model.compile(
            optimizer=Adam(learning_rate=0.0005),  # Slightly lower learning rate
            loss='mse',
            metrics=['mae']
        )

        print("✅ Model built successfully")
        print(f"Total parameters: {self.model.count_params():,}")

        return self.model.summary()

    def train_model(self, epochs=100, batch_size=32, patience=20, save_best=True):
        """
        Train the model with improved early stopping and learning rate scheduling
        """
        print(f"\n🚀 Training model for {epochs} epochs...")

        from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
        import os

        # Create models directory in Google Drive if it doesn't exist
        models_dir = '/content/drive/MyDrive/BearingRUL_Models'
        os.makedirs(models_dir, exist_ok=True)

        callbacks = [
            EarlyStopping(monitor='val_loss', patience=patience, restore_best_weights=True, verbose=1),
            ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=8, min_lr=0.00001, verbose=1)
        ]

        # Add model checkpoint to save best model to Google Drive
        if save_best:
            model_path = os.path.join(models_dir, f'best_model_{self.target_type}.h5')
            callbacks.append(
                ModelCheckpoint(
                    filepath=model_path,
                    monitor='val_loss',
                    save_best_only=True,
                    save_weights_only=False,
                    verbose=1
                )
            )
            print(f"Best model will be saved to Google Drive: {model_path}")

        # Check if we're using LSTM
        if hasattr(self, 'X_train_lstm'):
            X_train, X_val = self.X_train_lstm, self.X_val_lstm
        else:
            X_train, X_val = self.X_train_scaled, self.X_val_scaled

        history = self.model.fit(
            X_train, self.y_target_train,  # Use new target variable
            validation_data=(X_val, self.y_target_val),
            epochs=epochs,
            batch_size=batch_size,
            callbacks=callbacks,
            verbose=1
        )

        print("✅ Training completed")
        return history

    def evaluate_model(self):
        """
        Evaluate the trained model - FIXED BUG
        """
        if self.model is None:
            raise ValueError("Model not trained yet.")

        print("\n📊 Evaluating model performance...")

        # Get test data
        if hasattr(self, 'X_test_lstm'):
            X_test = self.X_test_lstm
        else:
            X_test = self.X_test_scaled

        # Predictions
        y_pred_target = self.model.predict(X_test, verbose=0).flatten()

        # RUL metrics - FIXED: Use correct target variable
        mse = mean_squared_error(self.y_target_test, y_pred_target)
        mae = mean_absolute_error(self.y_target_test, y_pred_target)
        rmse = np.sqrt(mse)

        # Convert predictions to RUL for health classification
        if self.target_type == 'health_percentage':
            # Convert percentage back to RUL hours
            max_rul = 128  # Original max RUL
            y_pred_rul = (y_pred_target / 100) * max_rul
            y_true_rul = (self.y_target_test / 100) * max_rul
        else:
            # Already in RUL format
            y_pred_rul = y_pred_target
            y_true_rul = self.y_target_test

        # Classification metrics (convert RUL to health status)
        def rul_to_health(rul):
            if rul > 80:
                return 0  # Healthy
            elif rul > 40:
                return 1  # Degrading
            elif rul > 15:
                return 2  # Near Failure
            else:
                return 3  # Critical

        y_pred_health = [rul_to_health(rul) for rul in y_pred_rul]
        y_true_health = [rul_to_health(rul) for rul in y_true_rul]

        accuracy = accuracy_score(y_true_health, y_pred_health)

        # Display target-specific metrics
        if self.target_type == 'health_percentage':
            print(f"📈 Health Percentage Prediction Metrics:")
            print(f"  MSE: {mse:.2f}")
            print(f"  MAE: {mae:.2f}%")
            print(f"  RMSE: {rmse:.2f}%")
        else:
            print(f"📈 RUL Prediction Metrics:")
            print(f"  MSE: {mse:.2f}")
            print(f"  MAE: {mae:.2f} hours")
            print(f"  RMSE: {rmse:.2f} hours")

        print(f"\n🎯 Health Classification Accuracy: {accuracy:.3f}")

        # Confusion Matrix
        cm = confusion_matrix(y_true_health, y_pred_health)

        plt.figure(figsize=(12, 5))

        # Plot 1: Target Predictions vs Actual
        plt.subplot(1, 2, 1)
        plt.scatter(self.y_target_test, y_pred_target, alpha=0.6)
        plt.plot([self.y_target_test.min(), self.y_target_test.max()],
                [self.y_target_test.min(), self.y_target_test.max()], 'r--', lw=2)

        if self.target_type == 'health_percentage':
            plt.xlabel('Actual Health Percentage (%)')
            plt.ylabel('Predicted Health Percentage (%)')
            plt.title(f'Health Percentage Prediction Results\nMAE: {mae:.2f}%')
        else:
            plt.xlabel('Actual RUL (hours)')
            plt.ylabel('Predicted RUL (hours)')
            plt.title(f'RUL Prediction Results\nMAE: {mae:.2f} hours')

        plt.grid(True, alpha=0.3)

        # Plot 2: Confusion Matrix
        plt.subplot(1, 2, 2)
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=['Healthy', 'Degrading', 'Near Failure', 'Critical'],
                   yticklabels=['Healthy', 'Degrading', 'Near Failure', 'Critical'])
        plt.title(f'Health Status Classification\nAccuracy: {accuracy:.3f}')
        plt.ylabel('True Status')
        plt.xlabel('Predicted Status')

        plt.tight_layout()
        plt.show()

        return {
            'mse': mse,
            'mae': mae,
            'rmse': rmse,
            'accuracy': accuracy,
            'confusion_matrix': cm
        }

    def predict_health(self, features):
        """
        Predict health status for new data

        Args:
            features: Feature vector or DataFrame
        """
        if self.model is None:
            raise ValueError("Model not trained yet.")

        # Ensure features are scaled
        if isinstance(features, pd.DataFrame):
            features_scaled = self.scaler.transform(features)
        else:
            features_scaled = self.scaler.transform(features.reshape(1, -1))

        # Reshape for LSTM if needed
        if hasattr(self, 'X_test_lstm'):
            features_scaled = features_scaled.reshape((features_scaled.shape[0], 1, features_scaled.shape[1]))

        # Predict target
        predicted_target = self.model.predict(features_scaled, verbose=0).flatten()

        # Convert to RUL if needed
        if self.target_type == 'health_percentage':
            predicted_rul = (predicted_target / 100) * 128  # Convert percentage to RUL
        else:
            predicted_rul = predicted_target

        # Convert to health status
        def rul_to_health_status(rul):
            if rul > 80:
                return "HEALTHY"
            elif rul > 40:
                return "DEGRADING"
            elif rul > 15:
                return "NEAR_FAILURE"
            else:
                return "CRITICAL"

        status = [rul_to_health_status(rul) for rul in predicted_rul]

        return predicted_rul, status

    def save_model_components(self, save_path='/content/drive/MyDrive/BearingRUL_Models'):
        """
        Save complete model with all components for deployment to Google Drive
        """
        import os
        import pickle
        from datetime import datetime
        from tensorflow.keras.models import save_model

        os.makedirs(save_path, exist_ok=True)
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

        # Save trained model
        model_path = os.path.join(save_path, f'trained_model_{timestamp}.keras')
        self.model.save(model_path)

        # Save scaler and metadata
        components_path = os.path.join(save_path, f'model_components_{timestamp}.pkl')
        components = {
            'scaler': self.scaler,
            'target_type': self.target_type,
            'feature_names': [col for col in self.features_df.columns
                            if col not in ['hour', 'window', 'rul', 'health_status', 'health_percentage']],
            'health_thresholds': {
                'healthy': 62.5,
                'degrading': 31.25,
                'near_failure': 11.7
            },
            'model_architecture': 'dense' if not hasattr(self, 'X_train_lstm') else 'lstm'
        }

        with open(components_path, 'wb') as f:
            pickle.dump(components, f)

        print(f"Model saved to Google Drive: {model_path}")
        print(f"Components saved to Google Drive: {components_path}")

        return model_path, components_path

    def load_trained_model(self, model_path, components_path):
        """
        Load complete trained model for deployment
        """
        import pickle
        from tensorflow.keras.models import load_model

        # Load model
        self.model = load_model(model_path)

        # Load components
        with open(components_path, 'rb') as f:
            components = pickle.load(f)

        self.scaler = components['scaler']
        self.target_type = components['target_type']
        self.feature_names = components['feature_names']
        self.health_thresholds = components['health_thresholds']

        print(f"Model loaded from: {model_path}")
        print(f"Components loaded from: {components_path}")
        print(f"Target type: {self.target_type}")
        print(f"Features: {len(self.feature_names)}")

        return True


def main():
    """
    Main execution function - loads preprocessed dataset and saves to new folder
    """
    import os
    from datetime import datetime

    # Path to the previously saved split dataset
    DATASET_PATH = "/content/drive/MyDrive/BearingRUL_SavedData/train_val_test_split_20250924_035736.pkl"  # Update with your actual path

    # Create new folder for this run
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    NEW_SAVE_PATH = f"/content/drive/MyDrive/BearingRUL_Models_{timestamp}"
    os.makedirs(NEW_SAVE_PATH, exist_ok=True)

    # Initialize the system without zip file (since we're loading preprocessed data)
    predictor = BearingRULPredictor(zip_file_path=None) # Pass None here

    try:
        # Step 1: Load preprocessed split dataset
        predictor.load_dataset(DATASET_PATH, stage='split')

        # Step 2: Build model
        predictor.build_model(model_type='lstm')  # Try 'lstm' for LSTM model

        # Step 3: Train model (will save best model to new folder)
        predictor.train_model(epochs=100, patience=20)

        # Step 4: Evaluate model
        results = predictor.evaluate_model()

        # Step 5: Save final model components to new folder
        model_path, components_path = predictor.save_model_components(save_path=NEW_SAVE_PATH)

        print("\n🎉 Training completed successfully!")
        print(f"Final MAE: {results['mae']:.2f}")
        print(f"Final Accuracy: {results['accuracy']:.3f}")
        print(f"\n💾 All files saved to new Google Drive folder: {NEW_SAVE_PATH}")
        print(f"  - Best model: {model_path}")
        print(f"  - Model components: {components_path}")

    except Exception as e:
        print(f"❌ Error: {e}")
        print("Please check your file paths and data format.")

if __name__ == "__main__":
    main()
````

````python
import pandas as pd
import numpy as np
import pickle
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import mean_squared_error, mean_absolute_error, accuracy_score, classification_report, confusion_matrix
from tensorflow.keras.models import load_model
import warnings
warnings.filterwarnings('ignore')

def test_trained_model(dataset_path, model_path, components_path, save_results=True, results_save_path=None):
    """
    Load saved train/test/val dataset and trained model, then test the model performance

    Args:
        dataset_path: Path to the saved train_val_test_split.pkl file
        model_path: Path to the trained model (.keras file)
        components_path: Path to the model components (.pkl file)
        save_results: Whether to save test results
        results_save_path: Path to save results (if None, uses same directory as model)

    Returns:
        dict: Test results including predictions and metrics
    """
    print("🚀 Loading dataset and trained model for testing...")

    # Step 1: Load the split dataset
    print(f"\n📁 Loading dataset from: {dataset_path}")
    try:
        with open(dataset_path, 'rb') as f:
            data = pickle.load(f)

        X_train_scaled = data['X_train']
        X_val_scaled = data['X_val']
        X_test_scaled = data['X_test']
        y_target_train = data['y_train']
        y_target_val = data['y_val']
        y_target_test = data['y_test']
        y_health_train = data['y_health_train']
        y_health_val = data['y_health_val']
        y_health_test = data['y_health_test']
        scaler = data['scaler']
        target_type = data['target_type']
        feature_names = data.get('feature_names', [])

        print(f"✅ Dataset loaded successfully:")
        print(f"  Training samples: {X_train_scaled.shape[0]}")
        print(f"  Validation samples: {X_val_scaled.shape[0]}")
        print(f"  Test samples: {X_test_scaled.shape[0]}")
        print(f"  Features: {X_test_scaled.shape[1]}")
        print(f"  Target type: {target_type}")

    except Exception as e:
        print(f"❌ Error loading dataset: {e}")
        return None

    # Step 2: Load the trained model
    print(f"\n🤖 Loading trained model from: {model_path}")
    try:
        model = load_model(model_path)
        print("✅ Model loaded successfully")
        print(f"  Model architecture: {len(model.layers)} layers")
        print(f"  Total parameters: {model.count_params():,}")
    except Exception as e:
        print(f"❌ Error loading model: {e}")
        return None

    # Step 3: Load model components
    print(f"\n⚙️ Loading model components from: {components_path}")
    try:
        with open(components_path, 'rb') as f:
            components = pickle.load(f)

        model_architecture = components.get('model_architecture', 'dense')
        health_thresholds = components.get('health_thresholds', {})

        print(f"✅ Components loaded successfully")
        print(f"  Architecture: {model_architecture}")

    except Exception as e:
        print(f"❌ Error loading components: {e}")
        return None

    # Step 4: Prepare test data based on model architecture
    print(f"\n🔄 Preparing test data for {model_architecture} model...")

    if model_architecture == 'lstm':
        # Reshape for LSTM (samples, timesteps, features)
        X_test_model = X_test_scaled.reshape((X_test_scaled.shape[0], 1, X_test_scaled.shape[1]))
        X_val_model = X_val_scaled.reshape((X_val_scaled.shape[0], 1, X_val_scaled.shape[1]))
        print("  Data reshaped for LSTM architecture")
    else:
        # Dense model uses data as-is
        X_test_model = X_test_scaled
        X_val_model = X_val_scaled
        print("  Using data for Dense architecture")

    # Step 5: Make predictions on test set
    print("\n🎯 Making predictions on test set...")

    try:
        y_pred_target = model.predict(X_test_model, verbose=0).flatten()
        print(f"✅ Predictions completed: {len(y_pred_target)} samples")

        # Also predict on validation set for comparison
        y_pred_val = model.predict(X_val_model, verbose=0).flatten()

    except Exception as e:
        print(f"❌ Error making predictions: {e}")
        return None

    # Step 6: Calculate metrics
    print("\n📊 Calculating performance metrics...")

    # Target prediction metrics (RUL or health percentage)
    test_mse = mean_squared_error(y_target_test, y_pred_target)
    test_mae = mean_absolute_error(y_target_test, y_pred_target)
    test_rmse = np.sqrt(test_mse)

    val_mse = mean_squared_error(y_target_val, y_pred_val)
    val_mae = mean_absolute_error(y_target_val, y_pred_val)
    val_rmse = np.sqrt(val_mse)

    # Convert predictions to RUL for health classification
    if target_type == 'health_percentage':
        max_rul = 128
        y_pred_rul = (y_pred_target / 100) * max_rul
        y_true_rul = (y_target_test / 100) * max_rul
        target_unit = "%"
    else:
        y_pred_rul = y_pred_target
        y_true_rul = y_target_test
        target_unit = "hours"

    # Health status classification
    def rul_to_health_status(rul):
        if rul > 80:
            return 0  # Healthy
        elif rul > 40:
            return 1  # Degrading
        elif rul > 15:
            return 2  # Near Failure
        else:
            return 3  # Critical

    y_pred_health = [rul_to_health_status(rul) for rul in y_pred_rul]
    y_true_health = [rul_to_health_status(rul) for rul in y_true_rul]

    # Classification metrics
    test_accuracy = accuracy_score(y_true_health, y_pred_health)
    cm = confusion_matrix(y_true_health, y_pred_health)
    class_report = classification_report(y_true_health, y_pred_health,
                                       target_names=['Healthy', 'Degrading', 'Near Failure', 'Critical'],
                                       output_dict=True)

    # Step 7: Display results
    print(f"\n🎯 TEST SET RESULTS:")
    print(f"=" * 50)
    print(f"📈 {target_type.replace('_', ' ').title()} Prediction Metrics:")
    print(f"  MSE: {test_mse:.4f}")
    print(f"  MAE: {test_mae:.2f} {target_unit}")
    print(f"  RMSE: {test_rmse:.2f} {target_unit}")

    print(f"\n🎯 Health Classification Metrics:")
    print(f"  Accuracy: {test_accuracy:.4f} ({test_accuracy*100:.1f}%)")

    print(f"\n📊 Validation vs Test Comparison:")
    print(f"  Validation MAE: {val_mae:.2f} {target_unit}")
    print(f"  Test MAE: {test_mae:.2f} {target_unit}")
    print(f"  Difference: {abs(test_mae - val_mae):.2f} {target_unit}")

    # Detailed classification report
    print(f"\n📋 Detailed Classification Report:")
    for class_name, metrics in class_report.items():
        if isinstance(metrics, dict) and class_name != 'accuracy':
            print(f"  {class_name}: Precision={metrics['precision']:.3f}, "
                  f"Recall={metrics['recall']:.3f}, F1={metrics['f1-score']:.3f}")

    # Step 8: Create visualizations
    print(f"\n📈 Creating visualizations...")

    plt.figure(figsize=(20, 12))

    # Plot 1: Predictions vs Actual (Test Set)
    plt.subplot(2, 4, 1)
    plt.scatter(y_target_test, y_pred_target, alpha=0.6, color='blue')
    plt.plot([y_target_test.min(), y_target_test.max()],
             [y_target_test.min(), y_target_test.max()], 'r--', lw=2)
    plt.xlabel(f'Actual {target_type.replace("_", " ").title()}')
    plt.ylabel(f'Predicted {target_type.replace("_", " ").title()}')
    plt.title(f'Test Set: Predictions vs Actual\nMAE: {test_mae:.2f} {target_unit}')
    plt.grid(True, alpha=0.3)

    # Plot 2: Confusion Matrix
    plt.subplot(2, 4, 2)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
               xticklabels=['Healthy', 'Degrading', 'Near Failure', 'Critical'],
               yticklabels=['Healthy', 'Degrading', 'Near Failure', 'Critical'])
    plt.title(f'Confusion Matrix\nAccuracy: {test_accuracy:.3f}')
    plt.ylabel('True Status')
    plt.xlabel('Predicted Status')

    # Plot 3: Residuals
    plt.subplot(2, 4, 3)
    residuals = y_target_test - y_pred_target
    plt.scatter(y_pred_target, residuals, alpha=0.6)
    plt.axhline(y=0, color='r', linestyle='--')
    plt.xlabel(f'Predicted {target_type.replace("_", " ").title()}')
    plt.ylabel('Residuals')
    plt.title('Residual Plot')
    plt.grid(True, alpha=0.3)

    # Plot 4: Error Distribution
    plt.subplot(2, 4, 4)
    plt.hist(residuals, bins=30, alpha=0.7, color='skyblue', edgecolor='black')
    plt.xlabel('Residuals')
    plt.ylabel('Frequency')
    plt.title(f'Error Distribution\nStd: {np.std(residuals):.2f}')
    plt.grid(True, alpha=0.3)

    # Plot 5: Validation vs Test Predictions
    plt.subplot(2, 4, 5)
    plt.scatter(y_target_val, y_pred_val, alpha=0.6, color='green', label='Validation')
    plt.scatter(y_target_test, y_pred_target, alpha=0.6, color='blue', label='Test')
    plt.plot([min(y_target_val.min(), y_target_test.min()),
              max(y_target_val.max(), y_target_test.max())],
             [min(y_target_val.min(), y_target_test.min()),
              max(y_target_val.max(), y_target_test.max())], 'r--', lw=2)
    plt.xlabel(f'Actual {target_type.replace("_", " ").title()}')
    plt.ylabel(f'Predicted {target_type.replace("_", " ").title()}')
    plt.title('Validation vs Test Performance')
    plt.legend()
    plt.grid(True, alpha=0.3)

    # Plot 6: Health Status Distribution
    plt.subplot(2, 4, 6)
    health_labels = ['Healthy', 'Degrading', 'Near Failure', 'Critical']
    true_counts = [sum(np.array(y_true_health) == i) for i in range(4)]
    pred_counts = [sum(np.array(y_pred_health) == i) for i in range(4)]

    x = np.arange(len(health_labels))
    width = 0.35

    plt.bar(x - width/2, true_counts, width, label='True', alpha=0.8)
    plt.bar(x + width/2, pred_counts, width, label='Predicted', alpha=0.8)
    plt.xlabel('Health Status')
    plt.ylabel('Count')
    plt.title('Health Status Distribution')
    plt.xticks(x, health_labels, rotation=45)
    plt.legend()
    plt.grid(True, alpha=0.3)

    # Plot 7: RUL Predictions (converted from target)
    plt.subplot(2, 4, 7)
    plt.scatter(y_true_rul, y_pred_rul, alpha=0.6, color='orange')
    plt.plot([y_true_rul.min(), y_true_rul.max()],
             [y_true_rul.min(), y_true_rul.max()], 'r--', lw=2)
    plt.xlabel('Actual RUL (hours)')
    plt.ylabel('Predicted RUL (hours)')
    plt.title(f'RUL Predictions\nMAE: {mean_absolute_error(y_true_rul, y_pred_rul):.2f} hours')
    plt.grid(True, alpha=0.3)

    # Plot 8: Prediction Confidence (sorted by error)
    plt.subplot(2, 4, 8)
    # Ensure residuals are a pandas Series for correct indexing
    residuals_series = pd.Series(residuals)
    abs_errors = np.abs(residuals_series)
    sorted_indices = np.argsort(abs_errors)
    # Use .iloc for positional indexing
    plt.plot(abs_errors.iloc[sorted_indices].values, alpha=0.7)
    plt.axhline(y=test_mae, color='r', linestyle='--', label=f'Mean AE: {test_mae:.2f}')
    plt.xlabel('Sample (sorted by error)')
    plt.ylabel('Absolute Error')
    plt.title('Prediction Errors (Sorted)')
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    # Step 9: Save results if requested
    results_dict = {
        'test_predictions': y_pred_target,
        'test_actual': y_target_test,
        'val_predictions': y_pred_val,
        'val_actual': y_target_val,
        'test_metrics': {
            'mse': test_mse,
            'mae': test_mae,
            'rmse': test_rmse,
            'accuracy': test_accuracy
        },
        'val_metrics': {
            'mse': val_mse,
            'mae': val_mae,
            'rmse': val_rmse
        },
        'confusion_matrix': cm,
        'classification_report': class_report,
        'target_type': target_type,
        'health_predictions': y_pred_health,
        'health_actual': y_true_health,
        'rul_predictions': y_pred_rul,
        'rul_actual': y_true_rul
    }

    if save_results:
        if results_save_path is None:
            results_save_path = model_path.replace('.keras', '_test_results.pkl') # Use .keras for new format

        try:
            with open(results_save_path, 'wb') as f:
                pickle.dump(results_dict, f)
            print(f"\n💾 Test results saved to: {results_save_path}")
        except Exception as e:
            print(f"⚠️ Could not save results: {e}")

    print(f"\n🎉 Testing completed successfully!")
    print(f"📊 Final Test Performance Summary:")
    print(f"  Target MAE: {test_mae:.2f} {target_unit}")
    print(f"  Health Classification Accuracy: {test_accuracy:.1%}")
    print(f"  RUL MAE: {mean_absolute_error(y_true_rul, y_pred_rul):.2f} hours")

    return results_dict


def test_model_example():
    """
    Example usage of the test_trained_model function
    Update the paths to match your saved files
    """
    # Update these paths to match your actual file locations
    DATASET_PATH = "/content/drive/MyDrive/BearingRUL_SavedData/train_val_test_split_20250924_035736.pkl"
    MODEL_PATH = "/content/drive/MyDrive/BearingRUL_Models_20250926_023323/trained_model_20250926_023409.keras"
    COMPONENTS_PATH = "/content/drive/MyDrive/BearingRUL_Models_20250926_023323/model_components_20250926_023409.pkl"

    # Run the test
    results = test_trained_model(
        dataset_path=DATASET_PATH,
        model_path=MODEL_PATH,
        components_path=COMPONENTS_PATH,
        save_results=True
    )

    if results:
        print("\n✅ Testing completed successfully!")
        return results
    else:
        print("\n❌ Testing failed. Please check your file paths and data.")
        return None


# Run the example (uncomment to use)
if __name__ == "__main__":
    test_model_example()
````

````python
import pandas as pd
import numpy as np
import pickle
import time
from datetime import datetime, timedelta
from tensorflow.keras.models import load_model
import warnings
warnings.filterwarnings('ignore')

class BearingHealthSimulator:
    def __init__(self, dataset_path, model_path, components_path):
        """
        Initialize the real-time bearing health monitoring simulator

        Args:
            dataset_path: Path to the processed dataset
            model_path: Path to the trained model
            components_path: Path to the model components
        """
        self.dataset_path = dataset_path
        self.model_path = model_path
        self.components_path = components_path

        # Simulation parameters
        self.simulation_speed = 2  # seconds between predictions
        self.current_index = 0
        self.simulation_history = []

        print("🏭 Bearing Health Monitoring Simulation Initializing...")
        self._load_components()

    def _load_components(self):
        """Load the model, dataset, and components"""
        try:
            # Load dataset
            print(f"📁 Loading dataset from: {self.dataset_path}")
            with open(self.dataset_path, 'rb') as f:
                data = pickle.load(f)

            # Reconstruct the full dataset chronologically
            X_all = np.vstack([data['X_train'], data['X_val'], data['X_test']])
            y_target_all = np.concatenate([data['y_train'], data['y_val'], data['y_test']])
            y_health_all = np.concatenate([data['y_health_train'], data['y_health_val'], data['y_health_test']])

            # Create DataFrame for easier handling
            feature_names = data.get('feature_names', [f'feature_{i}' for i in range(X_all.shape[1])])
            self.dataset = pd.DataFrame(X_all, columns=feature_names)
            self.dataset['target'] = y_target_all
            self.dataset['health_status'] = y_health_all

            # Sort by target (RUL or health_percentage) to simulate chronological degradation
            # Higher RUL/health_percentage = earlier in time (healthier state)
            self.dataset = self.dataset.sort_values('target', ascending=False).reset_index(drop=True)

            self.scaler = data['scaler']
            self.target_type = data['target_type']

            print(f"✅ Dataset loaded: {len(self.dataset)} data points")
            print(f"  Target type: {self.target_type}")

            # Load model
            print(f"🤖 Loading model from: {self.model_path}")
            self.model = load_model(self.model_path)

            # Load components
            with open(self.components_path, 'rb') as f:
                components = pickle.load(f)

            self.model_architecture = components.get('model_architecture', 'dense')
            self.health_thresholds = components.get('health_thresholds', {})

            print(f"✅ Model loaded successfully ({self.model_architecture} architecture)")
            print("🚀 Simulation ready to start!")

        except Exception as e:
            print(f"❌ Error loading components: {e}")
            raise

    def _prepare_input(self, features):
        """Prepare input data for the model"""
        # Get feature columns (exclude target and health_status)
        feature_cols = [col for col in features.columns if col not in ['target', 'health_status']]
        X = features[feature_cols].values

        # Reshape for LSTM if needed
        if self.model_architecture == 'lstm':
            X = X.reshape((X.shape[0], 1, X.shape[1]))

        return X

    def _convert_to_health_metrics(self, prediction, actual_target=None):
        """Convert model prediction to health metrics"""
        # Convert prediction to RUL if needed
        if self.target_type == 'health_percentage':
            predicted_rul = (prediction / 100) * 128  # Convert percentage to RUL hours
            if actual_target is not None:
                actual_rul = (actual_target / 100) * 128
            else:
                actual_rul = None
        else:
            predicted_rul = prediction
            actual_rul = actual_target

        # Determine health status from RUL
        def rul_to_status(rul):
            if rul > 80:
                return "HEALTHY", "🟢"
            elif rul > 40:
                return "DEGRADING", "🟡"
            elif rul > 15:
                return "NEAR_FAILURE", "🟠"
            else:
                return "CRITICAL", "🔴"

        pred_status, pred_icon = rul_to_status(predicted_rul)

        # Calculate confidence/severity
        max_rul = 128
        health_percentage = (predicted_rul / max_rul) * 100
        severity = 100 - health_percentage  # Higher severity = worse condition

        return {
            'predicted_rul': predicted_rul,
            'actual_rul': actual_rul,
            'health_status': pred_status,
            'status_icon': pred_icon,
            'health_percentage': health_percentage,
            'severity': severity,
            'raw_prediction': prediction
        }

    def _display_status(self, step, metrics, timestamp):
        """Display current bearing health status"""
        print("=" * 80)
        print(f"🏭 BEARING HEALTH MONITORING SYSTEM - Step {step}")
        print(f"⏰ Timestamp: {timestamp.strftime('%Y-%m-%d %H:%M:%S')}")
        print("-" * 80)

        # Main status display
        print(f"{metrics['status_icon']} STATUS: {metrics['health_status']}")
        print(f"📊 Health Percentage: {metrics['health_percentage']:.1f}%")
        print(f"⚠️  Severity Level: {metrics['severity']:.1f}%")
        print(f"⏳ Predicted RUL: {metrics['predicted_rul']:.1f} hours")

        if metrics['actual_rul'] is not None:
            print(f"🎯 Actual RUL: {metrics['actual_rul']:.1f} hours")
            error = abs(metrics['predicted_rul'] - metrics['actual_rul'])
            print(f"📏 Prediction Error: {error:.1f} hours")

        # Status-specific messages
        if metrics['health_status'] == "HEALTHY":
            print("✅ Machine operating normally. Continue regular monitoring.")
        elif metrics['health_status'] == "DEGRADING":
            print("⚠️  Machine showing signs of wear. Increase monitoring frequency.")
        elif metrics['health_status'] == "NEAR_FAILURE":
            print("🚨 Machine approaching failure. Schedule maintenance soon.")
        elif metrics['health_status'] == "CRITICAL":
            print("🔥 URGENT: Machine in critical condition. Immediate maintenance required!")

        # Trend analysis (if we have history)
        if len(self.simulation_history) > 1:
            prev_health = self.simulation_history[-2]['health_percentage']
            current_health = metrics['health_percentage']
            trend = current_health - prev_health

            if trend < -2:
                print(f"📉 TREND: Rapid degradation detected ({trend:.1f}% decline)")
            elif trend < -0.5:
                print(f"📉 TREND: Gradual degradation ({trend:.1f}% decline)")
            elif trend > 0.5:
                print(f"📈 TREND: Unexpected improvement ({trend:.1f}% increase)")
            else:
                print("📊 TREND: Stable condition")

        print("=" * 80)
        print()  # Empty line for spacing

    def _display_summary_stats(self):
        """Display summary statistics"""
        if len(self.simulation_history) < 2:
            return

        print("\n📈 SIMULATION SUMMARY STATISTICS")
        print("-" * 50)

        health_values = [h['health_percentage'] for h in self.simulation_history]
        rul_values = [h['predicted_rul'] for h in self.simulation_history]

        print(f"🔄 Total Steps Processed: {len(self.simulation_history)}")
        print(f"📊 Health Range: {min(health_values):.1f}% - {max(health_values):.1f}%")
        print(f"⏳ RUL Range: {min(rul_values):.1f} - {max(rul_values):.1f} hours")

        # Calculate degradation rate
        if len(health_values) > 1:
            total_degradation = health_values[0] - health_values[-1]
            steps = len(health_values) - 1
            degradation_rate = total_degradation / steps
            print(f"📉 Average Degradation Rate: {degradation_rate:.2f}% per step")

        # Status distribution
        statuses = [h['health_status'] for h in self.simulation_history]
        from collections import Counter
        status_counts = Counter(statuses)

        print(f"🟢 Healthy: {status_counts.get('HEALTHY', 0)} steps")
        print(f"🟡 Degrading: {status_counts.get('DEGRADING', 0)} steps")
        print(f"🟠 Near Failure: {status_counts.get('NEAR_FAILURE', 0)} steps")
        print(f"🔴 Critical: {status_counts.get('CRITICAL', 0)} steps")

    def run_simulation(self, max_steps=100, update_interval=2, show_summary=True):
        """
        Run the real-time bearing health monitoring simulation

        Args:
            max_steps: Maximum number of steps to simulate
            update_interval: Seconds between each prediction
            show_summary: Whether to show summary at the end
        """
        print(f"🚀 Starting Real-Time Bearing Health Monitoring Simulation")
        print(f"⚙️  Configuration:")
        print(f"   - Max Steps: {max_steps}")
        print(f"   - Update Interval: {update_interval} seconds")
        print(f"   - Dataset Size: {len(self.dataset)} samples")
        print(f"   - Model Architecture: {self.model_architecture}")
        print(f"\n🎬 Simulation starting in 3 seconds...\n")

        time.sleep(3)

        start_time = datetime.now()

        try:
            for step in range(1, max_steps + 1):
                # Check if we have more data
                if self.current_index >= len(self.dataset):
                    print("📋 Reached end of dataset. Simulation complete.")
                    break

                # Get current data point
                current_data = self.dataset.iloc[[self.current_index]]
                actual_target = current_data['target'].values[0]
                actual_health = current_data['health_status'].values[0]

                # Prepare input for model
                model_input = self._prepare_input(current_data)

                # Make prediction
                prediction = self.model.predict(model_input, verbose=0)[0][0]

                # Convert to health metrics
                metrics = self._convert_to_health_metrics(prediction, actual_target)
                metrics['actual_health_status'] = actual_health
                metrics['step'] = step
                metrics['timestamp'] = start_time + timedelta(seconds=(step-1) * update_interval)

                # Store in history
                self.simulation_history.append(metrics)

                # Display current status
                self._display_status(step, metrics, metrics['timestamp'])

                # Move to next data point
                self.current_index += 1

                # Wait for next update (except for last iteration)
                if step < max_steps and self.current_index < len(self.dataset):
                    time.sleep(update_interval)

        except KeyboardInterrupt:
            print(f"\n⏹️  Simulation stopped by user at step {step}")

        except Exception as e:
            print(f"\n❌ Error during simulation: {e}")

        finally:
            # Show summary
            if show_summary and self.simulation_history:
                self._display_summary_stats()

            print(f"\n🏁 Simulation completed after {len(self.simulation_history)} steps")
            print(f"⏰ Total runtime: {datetime.now() - start_time}")

    def run_fast_simulation(self, max_steps=50, show_every=5):
        """
        Run a fast simulation showing results every few steps

        Args:
            max_steps: Maximum number of steps
            show_every: Show results every N steps
        """
        print(f"⚡ Starting Fast Simulation (showing every {show_every} steps)")
        print("-" * 60)

        start_time = datetime.now()

        for step in range(1, max_steps + 1):
            if self.current_index >= len(self.dataset):
                break

            # Get current data and make prediction
            current_data = self.dataset.iloc[[self.current_index]]
            actual_target = current_data['target'].values[0]

            model_input = self._prepare_input(current_data)
            prediction = self.model.predict(model_input, verbose=0)[0][0]

            metrics = self._convert_to_health_metrics(prediction, actual_target)
            metrics['step'] = step

            self.simulation_history.append(metrics)

            # Display every N steps or at critical states
            if (step % show_every == 0 or
                metrics['health_status'] in ['NEAR_FAILURE', 'CRITICAL'] or
                step == 1 or step == max_steps):

                print(f"Step {step:3d}: {metrics['status_icon']} {metrics['health_status']:12s} | "
                      f"Health: {metrics['health_percentage']:5.1f}% | "
                      f"RUL: {metrics['predicted_rul']:6.1f}h | "
                      f"Severity: {metrics['severity']:5.1f}%")

                if metrics['health_status'] == 'CRITICAL':
                    print("         🚨 ALERT: Machine requires immediate attention! 🚨")

            self.current_index += 1

        print("-" * 60)
        print(f"⚡ Fast simulation completed: {len(self.simulation_history)} steps in {datetime.now() - start_time}")

        # Show final summary
        if self.simulation_history:
            final_metrics = self.simulation_history[-1]
            print(f"\n🏁 FINAL STATUS: {final_metrics['status_icon']} {final_metrics['health_status']}")
            print(f"   Health: {final_metrics['health_percentage']:.1f}%")
            print(f"   Predicted RUL: {final_metrics['predicted_rul']:.1f} hours")


def run_bearing_simulation():
    """
    Example function to run the bearing health simulation
    Update the paths to match your saved files
    """
    # Update these paths to match your actual file locations
    DATASET_PATH = "/content/drive/MyDrive/BearingRUL_SavedData/train_val_test_split_20250924_035736.pkl"
    MODEL_PATH = "/content/drive/MyDrive/BearingRUL_Models_20250926_023323/trained_model_20250926_023409.keras"
    COMPONENTS_PATH = "/content/drive/MyDrive/BearingRUL_Models_20250926_023323/model_components_20250926_023409.pkl"

    try:
        # Initialize simulator
        simulator = BearingHealthSimulator(
            dataset_path=DATASET_PATH,
            model_path=MODEL_PATH,
            components_path=COMPONENTS_PATH
        )

        print("\n🎮 Choose simulation mode:")
        print("1. Real-time simulation (2-3 second intervals)")
        print("2. Fast simulation (show every 5 steps)")

        # For Colab, we'll default to fast simulation to avoid long waits
        mode = input("\nEnter choice (1 or 2, default=2): ").strip() or "2"

        if mode == "1":
            # Real-time simulation
            max_steps = int(input("Enter max steps (default=30): ") or "30")
            interval = int(input("Enter update interval in seconds (default=3): ") or "3")

            print(f"\n🎬 Starting real-time simulation...")
            simulator.run_simulation(max_steps=max_steps, update_interval=interval)

        else:
            # Fast simulation
            max_steps = int(input("Enter max steps (default=100): ") or "100")
            show_every = int(input("Show every N steps (default=5): ") or "5")

            print(f"\n⚡ Starting fast simulation...")
            simulator.run_fast_simulation(max_steps=max_steps, show_every=show_every)

        print("\n✅ Simulation completed successfully!")
        return simulator

    except Exception as e:
        print(f"❌ Simulation failed: {e}")
        return None


# Auto-run simulation (uncomment to use)
if __name__ == "__main__":
    run_bearing_simulation()
````

````python
import pandas as pd
import numpy as np
import json
import pickle
from datetime import datetime
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    r2_score
)
from tensorflow.keras.models import load_model
import warnings
warnings.filterwarnings('ignore')

def generate_model_results_json(dataset_path, model_path, components_path, output_path=None):
    """
    Generate comprehensive JSON results for frontend visualization

    Args:
        dataset_path: Path to the saved train_val_test_split.pkl file
        model_path: Path to the trained model (.keras file)
        components_path: Path to the model components (.pkl file)
        output_path: Path to save JSON file (optional)

    Returns:
        dict: Comprehensive results dictionary
    """

    print("🔄 Generating comprehensive JSON results for frontend...")

    # Step 1: Load all components
    try:
        # Load dataset
        print("📁 Loading dataset...")
        with open(dataset_path, 'rb') as f:
            data = pickle.load(f)

        X_train_scaled = data['X_train']
        X_val_scaled = data['X_val']
        X_test_scaled = data['X_test']
        y_target_train = data['y_train']
        y_target_val = data['y_val']
        y_target_test = data['y_test']
        y_health_train = data['y_health_train']
        y_health_val = data['y_health_val']
        y_health_test = data['y_health_test']
        scaler = data['scaler']
        target_type = data['target_type']
        feature_names = data.get('feature_names', [])

        # Load model
        print("🤖 Loading model...")
        model = load_model(model_path)

        # Load components
        with open(components_path, 'rb') as f:
            components = pickle.load(f)

        model_architecture = components.get('model_architecture', 'dense')
        health_thresholds = components.get('health_thresholds', {})

        print("✅ All components loaded successfully")

    except Exception as e:
        print(f"❌ Error loading components: {e}")
        return None

    # Step 2: Prepare data for prediction
    def prepare_data_for_model(X_data):
        if model_architecture == 'lstm':
            return X_data.reshape((X_data.shape[0], 1, X_data.shape[1]))
        return X_data

    X_train_model = prepare_data_for_model(X_train_scaled)
    X_val_model = prepare_data_for_model(X_val_scaled)
    X_test_model = prepare_data_for_model(X_test_scaled)

    # Step 3: Make predictions
    print("🎯 Making predictions...")

    y_pred_train = model.predict(X_train_model, verbose=0).flatten()
    y_pred_val = model.predict(X_val_model, verbose=0).flatten()
    y_pred_test = model.predict(X_test_model, verbose=0).flatten()

    # Step 4: Convert predictions to different formats
    def convert_target_to_rul(target_values):
        """Convert target values to RUL hours"""
        if target_type == 'health_percentage':
            return (target_values / 100) * 128
        return target_values

    def rul_to_health_class(rul_values):
        """Convert RUL to health classification"""
        health_classes = []
        for rul in rul_values:
            if rul > 80:
                health_classes.append(0)  # Healthy
            elif rul > 40:
                health_classes.append(1)  # Degrading
            elif rul > 15:
                health_classes.append(2)  # Near Failure
            else:
                health_classes.append(3)  # Critical
        return np.array(health_classes)

    # Convert all predictions and actuals to RUL
    y_train_rul = convert_target_to_rul(y_target_train)
    y_val_rul = convert_target_to_rul(y_target_val)
    y_test_rul = convert_target_to_rul(y_target_test)

    y_pred_train_rul = convert_target_to_rul(y_pred_train)
    y_pred_val_rul = convert_target_to_rul(y_pred_val)
    y_pred_test_rul = convert_target_to_rul(y_pred_test)

    # Convert RUL to health classes for classification metrics
    y_train_health_pred = rul_to_health_class(y_pred_train_rul)
    y_val_health_pred = rul_to_health_class(y_pred_val_rul)
    y_test_health_pred = rul_to_health_class(y_pred_test_rul)

    # Step 5: Calculate all metrics for each dataset
    def calculate_comprehensive_metrics(y_true_target, y_pred_target, y_true_health, y_pred_health, dataset_name):
        """Calculate all metrics for a dataset"""

        # Regression metrics (for RUL/target prediction)
        mse = mean_squared_error(y_true_target, y_pred_target)
        mae = mean_absolute_error(y_true_target, y_pred_target)
        rmse = np.sqrt(mse)
        r2 = r2_score(y_true_target, y_pred_target)

        # Calculate percentage error
        # Avoid division by zero for MAPE
        mape_values = np.abs((y_true_target - y_pred_target) / y_true_target)
        mape = np.mean(mape_values[np.isfinite(mape_values)]) * 100 # Handle inf/NaN

        # Classification metrics (for health status)
        accuracy = accuracy_score(y_true_health, y_pred_health)

        # Handle multi-class classification metrics
        precision_macro = precision_score(y_true_health, y_pred_health, average='macro', zero_division=0)
        recall_macro = recall_score(y_true_health, y_pred_health, average='macro', zero_division=0)
        f1_macro = f1_score(y_true_health, y_pred_health, average='macro', zero_division=0)

        precision_weighted = precision_score(y_true_health, y_pred_health, average='weighted', zero_division=0)
        recall_weighted = recall_score(y_true_health, y_pred_health, average='weighted', zero_division=0)
        f1_weighted = f1_score(y_true_health, y_pred_health, average='weighted', zero_division=0)

        # Confusion Matrix
        cm = confusion_matrix(y_true_health, y_pred_health)

        # Classification report
        class_names = ['Healthy', 'Degrading', 'Near_Failure', 'Critical']
        class_report = classification_report(y_true_health, y_pred_health,
                                           target_names=class_names,
                                           output_dict=True,
                                           zero_division=0)

        return {
            'dataset': dataset_name,
            'sample_count': len(y_true_target),
            'regression_metrics': {
                'mse': round(float(mse), 4),
                'mae': round(float(mae), 4),
                'rmse': round(float(rmse), 4),
                'r2_score': round(float(r2), 4),
                'mape': round(float(mape), 4)
            },
            'classification_metrics': {
                'accuracy': round(float(accuracy), 4),
                'precision_macro': round(float(precision_macro), 4),
                'recall_macro': round(float(recall_macro), 4),
                'f1_macro': round(float(f1_macro), 4),
                'precision_weighted': round(float(precision_weighted), 4),
                'recall_weighted': round(float(recall_weighted), 4),
                'f1_weighted': round(float(f1_weighted), 4)
            },
            'confusion_matrix': {
                'matrix': cm.tolist(),
                'labels': class_names,
                # Avoid division by zero for normalized confusion matrix
                'normalized': (cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]).tolist() if cm.sum(axis=1).min() > 0 else cm.tolist()
            },
            'detailed_classification': {
                class_name: {
                    'precision': round(float(metrics['precision']), 4),
                    'recall': round(float(metrics['recall']), 4),
                    'f1_score': round(float(metrics['f1-score']), 4),
                    'support': int(metrics['support'])
                }
                for class_name, metrics in class_report.items()
                if class_name in class_names
            }
        }

    print("📊 Calculating comprehensive metrics...")

    # Calculate metrics for all datasets
    train_metrics = calculate_comprehensive_metrics(
        y_target_train, y_pred_train, y_health_train, y_train_health_pred, 'train'
    )

    val_metrics = calculate_comprehensive_metrics(
        y_target_val, y_pred_val, y_health_val, y_val_health_pred, 'validation'
    )

    test_metrics = calculate_comprehensive_metrics(
        y_target_test, y_pred_test, y_health_test, y_test_health_pred, 'test'
    )

    # Step 6: Prepare actual vs predicted data
    def prepare_predictions_data(y_true_target, y_pred_target, y_true_health, y_pred_health, dataset_name):
        """Prepare actual vs predicted data for visualization"""

        # Convert to NumPy arrays for consistent indexing
        y_true_target_np = np.array(y_true_target)
        y_pred_target_np = np.array(y_pred_target)
        y_true_health_np = np.array(y_true_health)
        y_pred_health_np = np.array(y_pred_health)

        # Convert to RUL for consistency
        y_true_rul = convert_target_to_rul(y_true_target_np)
        y_pred_rul = convert_target_to_rul(y_pred_target_np)


        predictions_data = []
        health_labels = ['Healthy', 'Degrading', 'Near_Failure', 'Critical']

        for i in range(len(y_true_target_np)):
            predictions_data.append({
                'index': int(i),
                'actual_target': round(float(y_true_target_np[i]), 4),
                'predicted_target': round(float(y_pred_target_np[i]), 4),
                'actual_rul': round(float(y_true_rul[i]), 4),
                'predicted_rul': round(float(y_pred_rul[i]), 4),
                'actual_health_class': int(y_true_health_np[i]),
                'predicted_health_class': int(y_pred_health_np[i]),
                'actual_health_label': health_labels[int(y_true_health_np[i])],
                'predicted_health_label': health_labels[int(y_pred_health_np[i])],
                'prediction_error': round(float(abs(y_true_target_np[i] - y_pred_target_np[i])), 4),
                'rul_error': round(float(abs(y_true_rul[i] - y_pred_rul[i])), 4),
                'classification_correct': bool(y_true_health_np[i] == y_pred_health_np[i])
            })

        return predictions_data

    print("📋 Preparing prediction data...")

    train_predictions = prepare_predictions_data(
        y_target_train, y_pred_train, y_health_train, y_train_health_pred, 'train'
    )

    val_predictions = prepare_predictions_data(
        y_target_val, y_pred_val, y_health_val, y_val_health_pred, 'validation'
    )

    test_predictions = prepare_predictions_data(
        y_target_test, y_pred_test, y_health_test, y_test_health_pred, 'test'
    )

    # Step 7: Create comprehensive results dictionary
    print("📦 Assembling final results...")

    results = {
        'metadata': {
            'generated_at': datetime.now().isoformat(),
            'model_architecture': model_architecture,
            'target_type': target_type,
            'total_features': len(feature_names),
            'feature_names': feature_names,
            'health_thresholds': health_thresholds,
            'class_labels': ['Healthy', 'Degrading', 'Near_Failure', 'Critical'],
            'dataset_sizes': {
                'train': len(y_target_train),
                'validation': len(y_target_val),
                'test': len(y_target_test),
                'total': len(y_target_train) + len(y_target_val) + len(y_target_test)
            }
        },
        'performance_metrics': {
            'train': train_metrics,
            'validation': val_metrics,
            'test': test_metrics
        },
        'predictions': {
            'train': train_predictions,
            'validation': val_predictions,
            'test': test_predictions
        },
        'summary_statistics': {
            'target_statistics': {
                'train': {
                    'actual': {
                        'mean': round(float(np.mean(y_target_train)), 4),
                        'std': round(float(np.std(y_target_train)), 4),
                        'min': round(float(np.min(y_target_train)), 4),
                        'max': round(float(np.max(y_target_train)), 4)
                    },
                    'predicted': {
                        'mean': round(float(np.mean(y_pred_train)), 4),
                        'std': round(float(np.std(y_pred_train)), 4),
                        'min': round(float(np.min(y_pred_train)), 4),
                        'max': round(float(np.max(y_pred_train)), 4)
                    }
                },
                'validation': {
                    'actual': {
                        'mean': round(float(np.mean(y_target_val)), 4),
                        'std': round(float(np.std(y_target_val)), 4),
                        'min': round(float(np.min(y_target_val)), 4),
                        'max': round(float(np.max(y_target_val)), 4)
                    },
                    'predicted': {
                        'mean': round(float(np.mean(y_pred_val)), 4),
                        'std': round(float(np.std(y_pred_val)), 4),
                        'min': round(float(np.min(y_pred_val)), 4),
                        'max': round(float(np.max(y_pred_val)), 4)
                    }
                },
                'test': {
                    'actual': {
                        'mean': round(float(np.mean(y_target_test)), 4),
                        'std': round(float(np.std(y_target_test)), 4),
                        'min': round(float(np.min(y_target_test)), 4),
                        'max': round(float(np.max(y_target_test)), 4)
                    },
                    'predicted': {
                        'mean': round(float(np.mean(y_pred_test)), 4),
                        'std': round(float(np.std(y_pred_test)), 4),
                        'min': round(float(np.min(y_pred_test)), 4),
                        'max': round(float(np.max(y_pred_test)), 4)
                    }
                }
            },
            'health_class_distribution': {
                'train': {
                    'actual': np.bincount(y_health_train, minlength=4).tolist(),
                    'predicted': np.bincount(y_train_health_pred, minlength=4).tolist()
                },
                'validation': {
                    'actual': np.bincount(y_health_val, minlength=4).tolist(),
                    'predicted': np.bincount(y_val_health_pred, minlength=4).tolist()
                },
                'test': {
                    'actual': np.bincount(y_health_test, minlength=4).tolist(),
                    'predicted': np.bincount(y_test_health_pred, minlength=4).tolist()
                }
            }
        }
    }

    # Step 8: Save to file if path provided
    if output_path:
        try:
            with open(output_path, 'w') as f:
                json.dump(results, f, indent=2)
            print(f"💾 Results saved to: {output_path}")
        except Exception as e:
            print(f"⚠️ Could not save to file: {e}")

    print("✅ JSON results generated successfully!")
    print(f"📊 Summary:")
    print(f"   - Test Accuracy: {results['performance_metrics']['test']['classification_metrics']['accuracy']:.4f}")
    print(f"   - Test MAE: {results['performance_metrics']['test']['regression_metrics']['mae']:.4f}")
    print(f"   - Test F1 Score: {results['performance_metrics']['test']['classification_metrics']['f1_weighted']:.4f}")

    return results


def generate_results_example():
    """
    Example function to generate JSON results
    Update the paths to match your saved files
    """
    # Update these paths to match your actual file locations
    DATASET_PATH = "/content/drive/MyDrive/BearingRUL_SavedData/train_val_test_split_20250924_035736.pkl"
    MODEL_PATH = "/content/drive/MyDrive/BearingRUL_Models_20250926_023323/trained_model_20250926_023409.keras"
    COMPONENTS_PATH = "/content/drive/MyDrive/BearingRUL_Models_20250926_023323/model_components_20250926_023409.pkl"
    OUTPUT_PATH = "/content/drive/MyDrive/BearingRUL_Results/model_results.json"

    # Generate results
    results = generate_model_results_json(
        dataset_path=DATASET_PATH,
        model_path=MODEL_PATH,
        components_path=COMPONENTS_PATH,
        output_path=OUTPUT_PATH
    )

    if results:
        print("\n✅ Results generated successfully!")

        # Display sample of the data structure
        print("\n📋 Sample JSON structure:")
        print(f"- Metadata: {len(results['metadata'])} fields")
        print(f"- Performance metrics: {len(results['performance_metrics'])} datasets")
        print(f"- Predictions: {len(results['predictions']['test'])} test samples")
        print(f"- Summary statistics: {len(results['summary_statistics'])} categories")

        return results
    else:
        print("\n❌ Failed to generate results")
        return None


# Usage example:
results = generate_results_example()

# To access specific data for frontend:
# test_accuracy = results['performance_metrics']['test']['classification_metrics']['accuracy']
# confusion_matrix = results['performance_metrics']['test']['confusion_matrix']['matrix']
# test_predictions = results['predictions']['test']
````

````python
import pandas as pd
import numpy as np
import zipfile
from scipy import stats
from scipy.signal import welch
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

class BearingDataProcessor:
    """
    Process bearing data and create chronological train/val/test splits
    """
    def __init__(self, zip_file_path):
        self.zip_file_path = zip_file_path
        self.raw_data = []
        self.features_df = None
        self.scaler = StandardScaler()
        print("🔧 Bearing Data Processor Initialized")
        print("⚠️  Will use CHRONOLOGICAL splitting")

    def load_and_clean_data(self):
        """Load data from zip file and perform initial cleaning"""
        print("\n📁 Loading and cleaning data from zip file...")

        with zipfile.ZipFile(self.zip_file_path, 'r') as zip_ref:
            file_list = sorted([f for f in zip_ref.namelist() if f.endswith('.csv')])
            print(f"Found {len(file_list)} CSV files")

            for idx, filename in enumerate(file_list):
                try:
                    with zip_ref.open(filename) as file:
                        df = pd.read_csv(file)

                    df = self._clean_single_file(df, idx)

                    self.raw_data.append({
                        'hour': idx,
                        'filename': filename,
                        'data': df,
                        'rul': 128 - idx
                    })

                    if idx % 20 == 0:
                        print(f"  Processed {idx+1}/{len(file_list)} files...")

                except Exception as e:
                    print(f"  ⚠️ Error processing {filename}: {e}")
                    continue

        print(f"✅ Successfully loaded {len(self.raw_data)} files")
        return len(self.raw_data)

    def _clean_single_file(self, df, hour):
        """Clean individual CSV file data"""
        expected_columns = ['vibration_x', 'vibration_y', 'temperature_bearing', 'temperature_atmospheric']

        if len(df.columns) == 4:
            df.columns = expected_columns

        df = df.dropna(how='all')

        for col in df.select_dtypes(include=[np.number]).columns:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

        df = df.fillna(method='ffill').fillna(method='bfill')
        return df

    def extract_features(self, window_size_seconds=10, overlap_ratio=0.5):
        """Extract features from raw signals"""
        print(f"\n🔍 Extracting features using {window_size_seconds}-second windows with {overlap_ratio*100}% overlap...")

        sampling_rate = 25600
        samples_per_window = sampling_rate * window_size_seconds
        step_size = int(samples_per_window * (1 - overlap_ratio))

        all_features = []

        for file_data in self.raw_data:
            hour = file_data['hour']
            df = file_data['data']
            rul = file_data['rul']

            total_samples = len(df)
            window_idx = 0

            for start_idx in range(0, total_samples - samples_per_window + 1, step_size):
                end_idx = start_idx + samples_per_window
                window_data = df.iloc[start_idx:end_idx]

                if len(window_data) < samples_per_window * 0.9:
                    continue

                features = self._extract_window_features(window_data, hour, window_idx, rul)
                all_features.append(features)
                window_idx += 1

        self.features_df = pd.DataFrame(all_features)
        self.features_df = self.features_df.sort_values('hour').reset_index(drop=True)

        print(f"✅ Extracted features: {self.features_df.shape[0]} windows × {self.features_df.shape[1]-3} features")
        print(f"📊 Time range: Hour {self.features_df['hour'].min()} to {self.features_df['hour'].max()}")

        # Free memory
        import gc
        del self.raw_data
        self.raw_data = []
        gc.collect()
        print("🗑️ Raw data freed from memory")

        return self.features_df.shape

    def _extract_window_features(self, window_data, hour, window_idx, rul):
        """Extract comprehensive features from a single window"""
        features = {'hour': hour, 'window': window_idx, 'rul': rul}

        for col in ['vibration_x', 'vibration_y', 'temperature_bearing', 'temperature_atmospheric']:
            if col in window_data.columns:
                signal = window_data[col].values
                features.update(self._time_domain_features(signal, col))
                if 'vibration' in col:
                    features.update(self._frequency_domain_features(signal, col))

        if 'vibration_x' in window_data.columns and 'vibration_y' in window_data.columns:
            features.update(self._cross_signal_features(
                window_data['vibration_x'].values,
                window_data['vibration_y'].values
            ))

        return features

    def _time_domain_features(self, signal, prefix):
        """Extract time-domain statistical features"""
        features = {}
        features[f'{prefix}_mean'] = np.mean(signal)
        features[f'{prefix}_std'] = np.std(signal)
        features[f'{prefix}_rms'] = np.sqrt(np.mean(signal**2))
        features[f'{prefix}_peak'] = np.max(np.abs(signal))
        features[f'{prefix}_peak_to_peak'] = np.ptp(signal)
        features[f'{prefix}_skewness'] = stats.skew(signal)
        features[f'{prefix}_kurtosis'] = stats.kurtosis(signal)
        features[f'{prefix}_variance'] = np.var(signal)

        rms_val = np.sqrt(np.mean(signal**2))
        features[f'{prefix}_crest_factor'] = np.max(np.abs(signal)) / rms_val if rms_val > 0 else 0
        mean_abs = np.mean(np.abs(signal))
        features[f'{prefix}_shape_factor'] = rms_val / mean_abs if mean_abs > 0 else 0

        return features

    def _frequency_domain_features(self, signal, prefix):
        """Extract frequency-domain features"""
        features = {}
        freqs, psd = welch(signal, fs=25600, nperseg=1024)

        features[f'{prefix}_spectral_centroid'] = np.sum(freqs * psd) / np.sum(psd)
        features[f'{prefix}_spectral_rolloff'] = freqs[np.where(np.cumsum(psd) >= 0.85 * np.sum(psd))[0][0]]

        low_freq_power = np.sum(psd[(freqs >= 0) & (freqs <= 500)])
        mid_freq_power = np.sum(psd[(freqs > 500) & (freqs <= 2000)])
        high_freq_power = np.sum(psd[freqs > 2000])

        total_power = np.sum(psd)
        features[f'{prefix}_low_freq_power'] = low_freq_power / total_power
        features[f'{prefix}_mid_freq_power'] = mid_freq_power / total_power
        features[f'{prefix}_high_freq_power'] = high_freq_power / total_power

        return features

    def _cross_signal_features(self, vib_x, vib_y):
        """Extract features from relationships between signals"""
        features = {}
        features['vib_xy_correlation'] = np.corrcoef(vib_x, vib_y)[0, 1]
        combined_vib = np.sqrt(vib_x**2 + vib_y**2)
        features['combined_vib_rms'] = np.sqrt(np.mean(combined_vib**2))
        features['combined_vib_peak'] = np.max(combined_vib)
        return features

    def create_health_labels(self):
        """Create health classification labels and percentage-based RUL"""
        if self.features_df is None:
            raise ValueError("Features not extracted yet. Run extract_features() first.")

        max_rul = self.features_df['rul'].max()
        self.features_df['health_percentage'] = (self.features_df['rul'] / max_rul) * 100

        def classify_health(health_pct):
            if health_pct > 62.5:
                return 0  # Healthy
            elif health_pct > 31.25:
                return 1  # Degrading
            elif health_pct > 11.7:
                return 2  # Near Failure
            else:
                return 3  # Critical

        self.features_df['health_status'] = self.features_df['health_percentage'].apply(classify_health)

        print("\n📊 Health Status Distribution:")
        print(self.features_df['health_status'].value_counts().sort_index())
        print(f"\n📊 Health Percentage Range: {self.features_df['health_percentage'].min():.1f}% - {self.features_df['health_percentage'].max():.1f}%")

        return self.features_df['health_status'].value_counts()

    def prepare_chronological_split(self, train_ratio=0.7, val_ratio=0.15, use_percentage=True):
        """
        Create CHRONOLOGICAL train/val/test split

        Args:
            train_ratio: Proportion for training (early hours)
            val_ratio: Proportion for validation (middle hours)
            use_percentage: Whether to use health_percentage instead of raw RUL
        """
        if self.features_df is None:
            raise ValueError("Features not extracted yet.")

        target_col = 'health_percentage' if use_percentage else 'rul'
        print(f"\n{'='*70}")
        print(f"🔄 CHRONOLOGICAL SPLIT - Time-based splitting for degradation data")
        print(f"{'='*70}")
        print(f"Target: {target_col}")

        # Ensure chronological order
        self.features_df = self.features_df.sort_values('hour').reset_index(drop=True)

        # Separate features from labels
        feature_columns = [col for col in self.features_df.columns
                          if col not in ['hour', 'window', 'rul', 'health_status', 'health_percentage']]

        X = self.features_df[feature_columns]
        y_target = self.features_df[target_col]
        y_health = self.features_df['health_status']
        hours = self.features_df['hour']

        # Handle NaN values
        X = X.fillna(X.mean())

        # CHRONOLOGICAL SPLIT based on hour
        max_hour = hours.max()
        train_cutoff = int(max_hour * train_ratio)
        val_cutoff = int(max_hour * (train_ratio + val_ratio))

        print(f"\n📅 Split Boundaries:")
        print(f"  Training:   Hour 0 → {train_cutoff} ({train_ratio*100:.0f}%)")
        print(f"  Validation: Hour {train_cutoff+1} → {val_cutoff} ({val_ratio*100:.0f}%)")
        print(f"  Testing:    Hour {val_cutoff+1} → {max_hour} ({(1-train_ratio-val_ratio)*100:.0f}%)")

        # Create masks for chronological split
        train_mask = hours <= train_cutoff
        val_mask = (hours > train_cutoff) & (hours <= val_cutoff)
        test_mask = hours > val_cutoff

        # Split data chronologically
        X_train = X[train_mask].values
        X_val = X[val_mask].values
        X_test = X[test_mask].values

        y_target_train = y_target[train_mask].values
        y_target_val = y_target[val_mask].values
        y_target_test = y_target[test_mask].values

        y_health_train = y_health[train_mask].values
        y_health_val = y_health[val_mask].values
        y_health_test = y_health[test_mask].values

        # Store hour boundaries
        train_hour_min, train_hour_max = hours[train_mask].min(), hours[train_mask].max()
        val_hour_min, val_hour_max = hours[val_mask].min(), hours[val_mask].max()
        test_hour_min, test_hour_max = hours[test_mask].min(), hours[test_mask].max()

        # Normalize features - FIT ONLY ON TRAINING DATA
        print(f"\n🔧 Normalizing features (fitted on training data only)...")
        X_train_scaled = self.scaler.fit_transform(X_train)
        X_val_scaled = self.scaler.transform(X_val)
        X_test_scaled = self.scaler.transform(X_test)

        print(f"\n✅ Chronological split completed:")
        print(f"  Training:   {X_train_scaled.shape[0]:>5} samples (Hours {train_hour_min:>3}-{train_hour_max:>3})")
        print(f"  Validation: {X_val_scaled.shape[0]:>5} samples (Hours {val_hour_min:>3}-{val_hour_max:>3})")
        print(f"  Testing:    {X_test_scaled.shape[0]:>5} samples (Hours {test_hour_min:>3}-{test_hour_max:>3})")
        print(f"  Features:   {X_train_scaled.shape[1]}")

        # Store all data
        self.split_data = {
            'X_train': X_train_scaled,
            'X_val': X_val_scaled,
            'X_test': X_test_scaled,
            'y_train': y_target_train,
            'y_val': y_target_val,
            'y_test': y_target_test,
            'y_health_train': y_health_train,
            'y_health_val': y_health_val,
            'y_health_test': y_health_test,
            'scaler': self.scaler,
            'target_type': target_col,
            'split_type': 'chronological',
            'split_info': {
                'train_hours': (int(train_hour_min), int(train_hour_max)),
                'val_hours': (int(val_hour_min), int(val_hour_max)),
                'test_hours': (int(test_hour_min), int(test_hour_max)),
                'train_ratio': train_ratio,
                'val_ratio': val_ratio,
                'test_ratio': 1 - train_ratio - val_ratio
            },
            'feature_names': feature_columns
        }

        print(f"\n✅ Data Leakage Check:")
        print(f"  ✓ Test set contains ONLY future hours (no overlap with training)")
        print(f"  ✓ Scaler fitted ONLY on training data")
        print(f"  ✓ No information from future leaked to past")

        return self.split_data

    def save_chronological_split(self, save_folder='/content/drive/MyDrive/BearingRUL_ChronologicalSplit'):
        """
        Save chronologically split data to a different folder

        Args:
            save_folder: Folder path (different from original splits)
        """
        import os
        import pickle
        from datetime import datetime

        if not hasattr(self, 'split_data'):
            raise ValueError("No split data available. Run prepare_chronological_split() first.")

        # Create folder if it doesn't exist
        os.makedirs(save_folder, exist_ok=True)

        # Generate timestamp
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

        # Create filename with clear naming
        filename = f'CHRONOLOGICAL_split_{timestamp}.pkl'
        filepath = os.path.join(save_folder, filename)

        # Save with all metadata
        print(f"\n💾 Saving chronological split...")
        with open(filepath, 'wb') as f:
            pickle.dump(self.split_data, f, protocol=pickle.HIGHEST_PROTOCOL)

        print(f"✅ Chronological split saved to:")
        print(f"   {filepath}")
        print(f"\n📋 File contains:")
        print(f"   • Training data: {self.split_data['X_train'].shape[0]} samples")
        print(f"   • Validation data: {self.split_data['X_val'].shape[0]} samples")
        print(f"   • Test data: {self.split_data['X_test'].shape[0]} samples")
        print(f"   • Features: {self.split_data['X_train'].shape[1]}")
        print(f"   • Split type: {self.split_data['split_type']}")
        print(f"   • Target type: {self.split_data['target_type']}")
        print(f"   • Hour boundaries: {self.split_data['split_info']}")

        return filepath


def main():
    """
    Main execution - Process data and create chronological split only
    """
    print("="*70)
    print("CHRONOLOGICAL DATA SPLIT PREPARATION")
    print("="*70)

    # Update this path to your zip file
    ZIP_FILE_PATH = "/content/drive/MyDrive/Vibration_Bearing_RuntoFailure.zip"

    # Update this to save in a DIFFERENT folder
    SAVE_FOLDER = "/content/drive/MyDrive/BearingRUL_ChronologicalSplit"

    try:
        # Initialize processor
        processor = BearingDataProcessor(ZIP_FILE_PATH)

        # Step 1: Load and clean data
        print("\n" + "="*70)
        print("STEP 1: Loading Data")
        print("="*70)
        processor.load_and_clean_data()

        # Step 2: Extract features
        print("\n" + "="*70)
        print("STEP 2: Extracting Features")
        print("="*70)
        processor.extract_features(window_size_seconds=10, overlap_ratio=0.5)

        # Step 3: Create health labels
        print("\n" + "="*70)
        print("STEP 3: Creating Health Labels")
        print("="*70)
        processor.create_health_labels()

        # Step 4: Create chronological split
        print("\n" + "="*70)
        print("STEP 4: Creating Chronological Split")
        print("="*70)
        processor.prepare_chronological_split(
            train_ratio=0.7,   # First 70% of time
            val_ratio=0.15,    # Next 15% of time
            use_percentage=True  # Use health_percentage as target
        )

        # Step 5: Save to different folder
        print("\n" + "="*70)
        print("STEP 5: Saving Split Data")
        print("="*70)
        filepath = processor.save_chronological_split(save_folder=SAVE_FOLDER)

        # Final summary
        print("\n" + "="*70)
        print("✅ PROCESS COMPLETED SUCCESSFULLY")
        print("="*70)
        print(f"📁 Chronological split saved to:")
        print(f"   {filepath}")
        print(f"\n🔄 Next step:")
        print(f"   Load this file for training with:")
        print(f"   import pickle")
        print(f"   with open('{filepath}', 'rb') as f:")
        print(f"       data = pickle.load(f)")
        print("="*70)

    except Exception as e:
        print(f"\n❌ Error occurred: {e}")
        import traceback
        traceback.print_exc()


if __name__ == "__main__":
    main()
````

````python
import pandas as pd
import numpy as np
import zipfile
import io
from scipy import stats
from scipy.signal import welch
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
import warnings
warnings.filterwarnings('ignore')
import pickle
from datetime import datetime

class BearingRULPredictor:
    def __init__(self, zip_file_path=None):
        """
        Initialize the Bearing RUL Prediction system

        Args:
            zip_file_path: Path to the zip file containing CSV files (optional if loading preprocessed data)
        """
        self.zip_file_path = zip_file_path
        self.raw_data = []
        self.features_df = None
        self.scaler = StandardScaler()
        self.model = None

        print("🔧 Bearing RUL Prediction System Initialized")

    def load_and_clean_data(self):
        """
        Load data from zip file and perform initial cleaning
        """
        print("\n📁 Loading and cleaning data from zip file...")

        with zipfile.Zipfile(self.zip_file_path, 'r') as zip_ref:
            file_list = sorted([f for f in zip_ref.namelist() if f.endswith('.csv')])
            print(f"Found {len(file_list)} CSV files")

            for idx, filename in enumerate(file_list):
                try:
                    # Read CSV from zip
                    with zip_ref.open(filename) as file:
                        df = pd.read_csv(file)

                    # Basic data cleaning
                    df = self._clean_single_file(df, idx)

                    # Store with metadata
                    self.raw_data.append({
                        'hour': idx,
                        'filename': filename,
                        'data': df,
                        'rul': 128 - idx  # RUL calculation
                    })

                    if idx % 20 == 0:
                        print(f"  Processed {idx+1}/{len(file_list)} files...")

                except Exception as e:
                    print(f"  ⚠️ Error processing {filename}: {e}")
                    continue

        print(f"✅ Successfully loaded {len(self.raw_data)} files")
        return len(self.raw_data)

    def _clean_single_file(self, df, hour):
        """
        Clean individual CSV file data

        Args:
            df: DataFrame from single CSV file
            hour: Hour index (0-128)
        """
        # Expected columns based on KAIST dataset description
        expected_columns = ['vibration_x', 'vibration_y', 'temperature_bearing', 'temperature_atmospheric']

        # Handle different possible column names
        if len(df.columns) == 4:
            df.columns = expected_columns

        # Remove any completely null rows
        df = df.dropna(how='all')

        # Handle outliers using IQR method
        for col in df.select_dtypes(include=[np.number]).columns:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)

        # Forward fill any remaining NaN values
        df = df.fillna(method='ffill')
        df = df.fillna(method='bfill')

        return df

    def extract_features(self, window_size_seconds=10, overlap_ratio=0.5):
        """
        Extract features from raw signals using overlapping sliding windows

        Args:
            window_size_seconds: Size of window for feature extraction
            overlap_ratio: Overlap ratio between consecutive windows (0-1)
        """
        print(f"\n🔍 Extracting features using {window_size_seconds}-second windows with {overlap_ratio*100}% overlap...")

        sampling_rate = 25600  # Hz
        samples_per_window = sampling_rate * window_size_seconds
        step_size = int(samples_per_window * (1 - overlap_ratio))  # Step size for overlapping

        all_features = []

        for file_data in self.raw_data:
            hour = file_data['hour']
            df = file_data['data']
            rul = file_data['rul']

            # Calculate overlapping windows
            total_samples = len(df)
            window_idx = 0

            for start_idx in range(0, total_samples - samples_per_window + 1, step_size):
                end_idx = start_idx + samples_per_window

                window_data = df.iloc[start_idx:end_idx]

                if len(window_data) < samples_per_window * 0.9:  # Skip windows that are too small
                    continue

                # Extract features for this window
                features = self._extract_window_features(window_data, hour, window_idx, rul)
                all_features.append(features)
                window_idx += 1

        self.features_df = pd.DataFrame(all_features)
        print(f"✅ Extracted features: {self.features_df.shape[0]} windows × {self.features_df.shape[1]-3} features")

        return self.features_df.shape

    def _extract_window_features(self, window_data, hour, window_idx, rul):
        """
        Extract comprehensive features from a single window
        """
        features = {
            'hour': hour,
            'window': window_idx,
            'rul': rul
        }

        # Process each sensor channel
        for col in ['vibration_x', 'vibration_y', 'temperature_bearing', 'temperature_atmospheric']:
            if col in window_data.columns:
                signal = window_data[col].values

                # Time domain features
                features.update(self._time_domain_features(signal, col))

                # Frequency domain features (only for vibration)
                if 'vibration' in col:
                    features.update(self._frequency_domain_features(signal, col))

        # Cross-signal features
        if 'vibration_x' in window_data.columns and 'vibration_y' in window_data.columns:
            features.update(self._cross_signal_features(
                window_data['vibration_x'].values,
                window_data['vibration_y'].values
            ))

        return features

    def _time_domain_features(self, signal, prefix):
        """Extract time-domain statistical features"""
        features = {}

        features[f'{prefix}_mean'] = np.mean(signal)
        features[f'{prefix}_std'] = np.std(signal)
        features[f'{prefix}_rms'] = np.sqrt(np.mean(signal**2))
        features[f'{prefix}_peak'] = np.max(np.abs(signal))
        features[f'{prefix}_peak_to_peak'] = np.ptp(signal)
        features[f'{prefix}_skewness'] = stats.skew(signal)
        features[f'{prefix}_kurtosis'] = stats.kurtosis(signal)
        features[f'{prefix}_variance'] = np.var(signal)

        # Crest factor
        rms_val = np.sqrt(np.mean(signal**2))
        features[f'{prefix}_crest_factor'] = np.max(np.abs(signal)) / rms_val if rms_val > 0 else 0

        # Shape factor
        mean_abs = np.mean(np.abs(signal))
        features[f'{prefix}_shape_factor'] = rms_val / mean_abs if mean_abs > 0 else 0

        return features

    def _frequency_domain_features(self, signal, prefix):
        """Extract frequency-domain features"""
        features = {}

        # Power Spectral Density
        freqs, psd = welch(signal, fs=25600, nperseg=1024)

        # Spectral features
        features[f'{prefix}_spectral_centroid'] = np.sum(freqs * psd) / np.sum(psd)
        features[f'{prefix}_spectral_rolloff'] = freqs[np.where(np.cumsum(psd) >= 0.85 * np.sum(psd))[0][0]]

        # Power in frequency bands (bearing fault frequencies)
        low_freq_power = np.sum(psd[(freqs >= 0) & (freqs <= 500)])
        mid_freq_power = np.sum(psd[(freqs > 500) & (freqs <= 2000)])
        high_freq_power = np.sum(psd[freqs > 2000])

        total_power = np.sum(psd)
        features[f'{prefix}_low_freq_power'] = low_freq_power / total_power
        features[f'{prefix}_mid_freq_power'] = mid_freq_power / total_power
        features[f'{prefix}_high_freq_power'] = high_freq_power / total_power

        return features

    def _cross_signal_features(self, vib_x, vib_y):
        """Extract features from relationships between signals"""
        features = {}

        # Correlation between vibration axes
        features['vib_xy_correlation'] = np.corrcoef(vib_x, vib_y)[0, 1]

        # Combined vibration magnitude
        combined_vib = np.sqrt(vib_x**2 + vib_y**2)
        features['combined_vib_rms'] = np.sqrt(np.mean(combined_vib**2))
        features['combined_vib_peak'] = np.max(combined_vib)

        return features

    def create_health_labels(self):
        """
        Create health classification labels and percentage-based RUL
        """
        if self.features_df is None:
            raise ValueError("Features not extracted yet. Run extract_features() first.")

        # Create percentage-based health indicator
        max_rul = self.features_df['rul'].max()  # 128 hours
        self.features_df['health_percentage'] = (self.features_df['rul'] / max_rul) * 100

        def classify_health(health_pct):
            if health_pct > 62.5:  # > 80 hours (62.5% of 128)
                return 0  # Healthy
            elif health_pct > 31.25:  # > 40 hours (31.25% of 128)
                return 1  # Degrading
            elif health_pct > 11.7:  # > 15 hours (11.7% of 128)
                return 2  # Near Failure
            else:
                return 3  # Critical

        self.features_df['health_status'] = self.features_df['health_percentage'].apply(classify_health)

        print("\n📊 Health Status Distribution:")
        print(self.features_df['health_status'].value_counts().sort_index())
        print(f"\n📊 Health Percentage Range: {self.features_df['health_percentage'].min():.1f}% - {self.features_df['health_percentage'].max():.1f}%")

        return self.features_df['health_status'].value_counts()

    def save_dataset(self, stage='features', save_path='/content/drive/MyDrive/BearingRUL_SavedData'):
        """
        Save dataset at different processing stages to Google Drive

        Args:
            stage: 'raw', 'features', 'processed', or 'split'
            save_path: Directory to save data (default: Google Drive)
        """
        import os
        import pickle
        from datetime import datetime
        import gc

        os.makedirs(save_path, exist_ok=True)
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

        if stage == 'raw':
            # Skip saving raw data to save RAM - it's too large
            print("⚠️ Skipping raw data save to prevent RAM crash - data is too large")
            print("✅ Raw data processing completed (not saved to preserve memory)")
            return None

        elif stage == 'features':
            # Save feature dataframe and free up raw data memory
            filename = f'features_dataset_{timestamp}.csv'
            filepath = os.path.join(save_path, filename)
            self.features_df.to_csv(filepath, index=False)
            print(f"✅ Features dataset saved to Google Drive: {filepath}")

            # Free up raw data memory after feature extraction
            if hasattr(self, 'raw_data') and self.raw_data:
                print("🗑️ Freeing raw data from memory to save RAM...")
                del self.raw_data
                self.raw_data = []
                gc.collect()
                print("✅ Memory freed")

        elif stage == 'processed':
            # Save features with labels
            filename = f'processed_dataset_{timestamp}.pkl'
            filepath = os.path.join(save_path, filename)
            processed_data = {
                'features_df': self.features_df,
                'target_type': getattr(self, 'target_type', 'rul'),
                'scaler': self.scaler if hasattr(self, 'scaler') else None
            }

            # Use protocol 4 for better memory efficiency
            with open(filepath, 'wb') as f:
                pickle.dump(processed_data, f, protocol=pickle.HIGHEST_PROTOCOL)
            print(f"✅ Processed dataset saved to Google Drive: {filepath}")

            # Clean up temporary variables
            del processed_data
            gc.collect()

        elif stage == 'split':
            # Save train/val/test splits
            filename = f'train_val_test_split_{timestamp}.pkl'
            filepath = os.path.join(save_path, filename)
            split_data = {
                'X_train': self.X_train_scaled,
                'X_val': self.X_val_scaled,
                'X_test': self.X_test_scaled,
                'y_train': self.y_target_train,
                'y_val': self.y_target_val,
                'y_test': self.y_target_test,
                'y_health_train': self.y_health_train,
                'y_health_val': self.y_health_val,
                'y_health_test': self.y_health_test,
                'scaler': self.scaler,
                'target_type': self.target_type,
                # Include feature_names for reconstruction
                'feature_names': feature_columns # Use the local feature_columns here
            }

            with open(filepath, 'wb') as f:
                pickle.dump(split_data, f)
            print(f"✅ Train/val/test splits saved to Google Drive: {filepath}")

        return filepath

    def load_dataset(self, filepath, stage='split'):
        """
        Load previously saved dataset

        Args:
            filepath: Path to saved data file
            stage: 'raw', 'features', 'processed', or 'split'
        """
        import pickle

        if stage == 'raw':
            with open(filepath, 'rb') as f:
                self.raw_data = pickle.load(f)
            print(f"✅ Raw data loaded from: {filepath}")

        elif stage == 'features':
            self.features_df = pd.read_csv(filepath)
            print(f"✅ Features loaded from: {filepath}")

        elif stage == 'processed':
            with open(filepath, 'rb') as f:
                data = pickle.load(f)
            self.features_df = data['features_df']
            self.target_type = data.get('target_type', 'rul')
            if data['scaler']:
                self.scaler = data['scaler']
            print(f"✅ Processed data loaded from: {filepath}")

        elif stage == 'split':
            with open(filepath, 'rb') as f:
                data = pickle.load(f)

            self.X_train_scaled = data['X_train']
            self.X_val_scaled = data['X_val']
            self.X_test_scaled = data['X_test']
            self.y_target_train = data['y_train']
            self.y_target_val = data['y_val']
            self.y_target_test = data['y_test']
            self.y_health_train = data['y_health_train']
            self.y_health_val = data['y_health_val']
            self.y_health_test = data['y_health_test']
            self.scaler = data['scaler']
            self.target_type = data['target_type']

            # Reconstruct features_df from split data
            feature_names = data['feature_names']
            # Concatenate scaled data
            all_scaled_X = np.vstack([self.X_train_scaled, self.X_val_scaled, self.X_test_scaled])
            self.features_df = pd.DataFrame(all_scaled_X, columns=feature_names)

            # Add back metadata columns (approximate reconstruction)
            all_y_target = np.concatenate([self.y_target_train, self.y_target_val, self.y_target_test])
            all_y_health = np.concatenate([self.y_health_train, self.y_health_val, self.y_health_test])

            if self.target_type == 'health_percentage':
                self.features_df['health_percentage'] = all_y_target
                # Approximate RUL from health percentage
                self.features_df['rul'] = (all_y_target / 100) * 128
            else: # target_type is 'rul'
                self.features_df['rul'] = all_y_target
                # Approximate health percentage from RUL
                max_rul = 128 # Assuming original max RUL was 128
                self.features_df['health_percentage'] = (all_y_target / max_rul) * 100

            self.features_df['health_status'] = all_y_health
            # Add hour and window - these cannot be perfectly reconstructed from split data alone,
            # so we'll add placeholder values or attempt a simple approximation if the original
            # structure (e.g., sorted by hour/window) is preserved in the split.
            # For now, adding simple index as placeholder. A more robust approach would save
            # hour/window in the split data if needed for saving features_df.
            # Let's add placeholder columns to avoid errors in save_model_components
            if 'hour' not in self.features_df.columns:
                 self.features_df['hour'] = -1 # Placeholder
            if 'window' not in self.features_df.columns:
                 self.features_df['window'] = range(len(self.features_df)) # Placeholder


            print(f"✅ Train/val/test splits loaded from: {filepath}")
            print(f"  Training: {self.X_train_scaled.shape[0]} samples")
            print(f"  Validation: {self.X_val_scaled.shape[0]} samples")
            print(f"  Testing: {self.X_test_scaled.shape[0]} samples")

        return True

    def prepare_data(self, test_size=0.2, val_size=0.1, use_percentage=True):
        """
        Normalize features and split dataset

        Args:
            test_size: Proportion for test set
            val_size: Proportion for validation set
            use_percentage: Whether to use health_percentage instead of raw RUL
        """
        if self.features_df is None:
            raise ValueError("Features not extracted yet.")

        target_col = 'health_percentage' if use_percentage else 'rul'
        print(f"\n🔄 Preparing data with {test_size*100}% test, {val_size*100}% validation split...")
        print(f"Using target: {target_col}")

        # Separate features from labels
        feature_columns = [col for col in self.features_df.columns
                          if col not in ['hour', 'window', 'rul', 'health_status', 'health_percentage']]

        X = self.features_df[feature_columns]
        y_target = self.features_df[target_col]
        y_health = self.features_df['health_status']

        # Handle any remaining NaN values
        X = X.fillna(X.mean())

        # Split the data
        X_temp, self.X_test, y_target_temp, self.y_target_test, y_health_temp, self.y_health_test = \
            train_test_split(X, y_target, y_health, test_size=test_size, random_state=42, stratify=y_health)

        val_size_adjusted = val_size / (1 - test_size)
        self.X_train, self.X_val, self.y_target_train, self.y_target_val, self.y_health_train, self.y_health_val = \
            train_test_split(X_temp, y_target_temp, y_health_temp, test_size=val_size_adjusted,
                           random_state=42, stratify=y_health_temp)

        # Store which target we're using
        self.target_type = target_col

        # Normalize features
        self.X_train_scaled = self.scaler.fit_transform(self.X_train)
        self.X_val_scaled = self.scaler.transform(self.X_val)
        self.X_test_scaled = self.scaler.transform(self.X_test)

        print(f"✅ Data prepared:")
        print(f"  Training: {self.X_train_scaled.shape[0]} samples")
        print(f"  Validation: {self.X_val_scaled.shape[0]} samples")
        print(f"  Testing: {self.X_test_scaled.shape[0]} samples")
        print(f"  Features: {self.X_train_scaled.shape[1]}")

        return {
            'train_shape': self.X_train_scaled.shape,
            'val_shape': self.X_val_scaled.shape,
            'test_shape': self.X_test_scaled.shape
        }

    def build_model(self, model_type='dense'):
        """
        Build improved neural network model for RUL prediction

        Args:
            model_type: 'dense' or 'lstm'
        """
        print(f"\n🏗️ Building improved {model_type.upper()} model...")

        if model_type == 'dense':
            self.model = Sequential([
                Dense(256, activation='relu', input_shape=(self.X_train_scaled.shape[1],)),
                Dropout(0.4),
                Dense(128, activation='relu'),
                Dropout(0.3),
                Dense(64, activation='relu'),
                Dropout(0.2),
                Dense(32, activation='relu'),
                Dropout(0.1),
                Dense(1, activation='linear')  # RUL/percentage prediction
            ])

        elif model_type == 'lstm':
            # Reshape data for LSTM (samples, timesteps, features)
            self.X_train_lstm = self.X_train_scaled.reshape((self.X_train_scaled.shape[0], 1, self.X_train_scaled.shape[1]))
            self.X_val_lstm = self.X_val_scaled.reshape((self.X_val_scaled.shape[0], 1, self.X_val_scaled.shape[1]))
            self.X_test_lstm = self.X_test_scaled.reshape((self.X_test_scaled.shape[0], 1, self.X_test_scaled.shape[1]))

            self.model = Sequential([
                LSTM(128, return_sequences=True, input_shape=(1, self.X_train_scaled.shape[1])),
                Dropout(0.4),
                LSTM(64, return_sequences=False),
                Dropout(0.3),
                Dense(32, activation='relu'),
                Dropout(0.2),
                Dense(1, activation='linear')
            ])

        # Compile model with improved settings
        self.model.compile(
            optimizer=Adam(learning_rate=0.0005),  # Slightly lower learning rate
            loss='mse',
            metrics=['mae']
        )

        print("✅ Model built successfully")
        print(f"Total parameters: {self.model.count_params():,}")

        return self.model.summary()

    def train_model(self, epochs=100, batch_size=32, patience=20, save_best=True):
        """
        Train the model with improved early stopping and learning rate scheduling
        """
        print(f"\n🚀 Training model for {epochs} epochs...")

        from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
        import os

        # Create models directory in Google Drive if it doesn't exist
        models_dir = '/content/drive/MyDrive/BearingRUL_Models'
        os.makedirs(models_dir, exist_ok=True)

        callbacks = [
            EarlyStopping(monitor='val_loss', patience=patience, restore_best_weights=True, verbose=1),
            ReduceLROnPlateau(monitor='val_loss', factor=0.3, patience=8, min_lr=0.00001, verbose=1)
        ]

        # Add model checkpoint to save best model to Google Drive
        if save_best:
            model_path = os.path.join(models_dir, f'best_model_{self.target_type}.h5')
            callbacks.append(
                ModelCheckpoint(
                    filepath=model_path,
                    monitor='val_loss',
                    save_best_only=True,
                    save_weights_only=False,
                    verbose=1
                )
            )
            print(f"Best model will be saved to Google Drive: {model_path}")

        # Check if we're using LSTM
        if hasattr(self, 'X_train_lstm'):
            X_train, X_val = self.X_train_lstm, self.X_val_lstm
        else:
            X_train, X_val = self.X_train_scaled, self.X_val_scaled

        history = self.model.fit(
            X_train, self.y_target_train,  # Use new target variable
            validation_data=(X_val, self.y_target_val),
            epochs=epochs,
            batch_size=batch_size,
            callbacks=callbacks,
            verbose=1
        )

        print("✅ Training completed")
        return history

    def evaluate_model(self):
        """
        Evaluate the trained model - FIXED BUG
        """
        if self.model is None:
            raise ValueError("Model not trained yet.")

        print("\n📊 Evaluating model performance...")

        # Get test data
        if hasattr(self, 'X_test_lstm'):
            X_test = self.X_test_lstm
        else:
            X_test = self.X_test_scaled

        # Predictions
        y_pred_target = self.model.predict(X_test, verbose=0).flatten()

        # RUL metrics - FIXED: Use correct target variable
        mse = mean_squared_error(self.y_target_test, y_pred_target)
        mae = mean_absolute_error(self.y_target_test, y_pred_target)
        rmse = np.sqrt(mse)

        # Convert predictions to RUL for health classification
        if self.target_type == 'health_percentage':
            # Convert percentage back to RUL hours
            max_rul = 128  # Original max RUL
            y_pred_rul = (y_pred_target / 100) * max_rul
            y_true_rul = (self.y_target_test / 100) * max_rul
        else:
            # Already in RUL format
            y_pred_rul = y_pred_target
            y_true_rul = self.y_target_test

        # Classification metrics (convert RUL to health status)
        def rul_to_health(rul):
            if rul > 80:
                return 0  # Healthy
            elif rul > 40:
                return 1  # Degrading
            elif rul > 15:
                return 2  # Near Failure
            else:
                return 3  # Critical

        y_pred_health = [rul_to_health(rul) for rul in y_pred_rul]
        y_true_health = [rul_to_health(rul) for rul in y_true_rul]

        accuracy = accuracy_score(y_true_health, y_pred_health)

        # Display target-specific metrics
        if self.target_type == 'health_percentage':
            print(f"📈 Health Percentage Prediction Metrics:")
            print(f"  MSE: {mse:.2f}")
            print(f"  MAE: {mae:.2f}%")
            print(f"  RMSE: {rmse:.2f}%")
        else:
            print(f"📈 RUL Prediction Metrics:")
            print(f"  MSE: {mse:.2f}")
            print(f"  MAE: {mae:.2f} hours")
            print(f"  RMSE: {rmse:.2f} hours")

        print(f"\n🎯 Health Classification Accuracy: {accuracy:.3f}")

        # Confusion Matrix
        cm = confusion_matrix(y_true_health, y_pred_health)

        plt.figure(figsize=(12, 5))

        # Plot 1: Target Predictions vs Actual
        plt.subplot(1, 2, 1)
        plt.scatter(self.y_target_test, y_pred_target, alpha=0.6)
        plt.plot([self.y_target_test.min(), self.y_target_test.max()],
                [self.y_target_test.min(), self.y_target_test.max()], 'r--', lw=2)

        if self.target_type == 'health_percentage':
            plt.xlabel('Actual Health Percentage (%)')
            plt.ylabel('Predicted Health Percentage (%)')
            plt.title(f'Health Percentage Prediction Results\nMAE: {mae:.2f}%')
        else:
            plt.xlabel('Actual RUL (hours)')
            plt.ylabel('Predicted RUL (hours)')
            plt.title(f'RUL Prediction Results\nMAE: {mae:.2f} hours')

        plt.grid(True, alpha=0.3)

        # Plot 2: Confusion Matrix
        plt.subplot(1, 2, 2)
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=['Healthy', 'Degrading', 'Near Failure', 'Critical'],
                   yticklabels=['Healthy', 'Degrading', 'Near Failure', 'Critical'])
        plt.title(f'Health Status Classification\nAccuracy: {accuracy:.3f}')
        plt.ylabel('True Status')
        plt.xlabel('Predicted Status')

        plt.tight_layout()
        plt.show()

        return {
            'mse': mse,
            'mae': mae,
            'rmse': rmse,
            'accuracy': accuracy,
            'confusion_matrix': cm
        }

    def predict_health(self, features):
        """
        Predict health status for new data

        Args:
            features: Feature vector or DataFrame
        """
        if self.model is None:
            raise ValueError("Model not trained yet.")

        # Ensure features are scaled
        if isinstance(features, pd.DataFrame):
            features_scaled = self.scaler.transform(features)
        else:
            features_scaled = self.scaler.transform(features.reshape(1, -1))

        # Reshape for LSTM if needed
        if hasattr(self, 'X_test_lstm'):
            features_scaled = features_scaled.reshape((features_scaled.shape[0], 1, features_scaled.shape[1]))

        # Predict target
        predicted_target = self.model.predict(features_scaled, verbose=0).flatten()

        # Convert to RUL if needed
        if self.target_type == 'health_percentage':
            predicted_rul = (predicted_target / 100) * 128  # Convert percentage to RUL
        else:
            predicted_rul = predicted_target

        # Convert to health status
        def rul_to_health_status(rul):
            if rul > 80:
                return "HEALTHY"
            elif rul > 40:
                return "DEGRADING"
            elif rul > 15:
                return "NEAR_FAILURE"
            else:
                return "CRITICAL"

        status = [rul_to_health_status(rul) for rul in predicted_rul]

        return predicted_rul, status

    def save_model_components(self, save_path='/content/drive/MyDrive/BearingRUL_Models'):
        """
        Save complete model with all components for deployment to Google Drive
        """
        import os
        import pickle
        from datetime import datetime
        from tensorflow.keras.models import save_model

        os.makedirs(save_path, exist_ok=True)
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

        # Save trained model
        model_path = os.path.join(save_path, f'trained_model_{timestamp}.keras')
        self.model.save(model_path)

        # Save scaler and metadata
        components_path = os.path.join(save_path, f'model_components_{timestamp}.pkl')
        components = {
            'scaler': self.scaler,
            'target_type': self.target_type,
            'feature_names': [col for col in self.features_df.columns
                            if col not in ['hour', 'window', 'rul', 'health_status', 'health_percentage']],
            'health_thresholds': {
                'healthy': 62.5,
                'degrading': 31.25,
                'near_failure': 11.7
            },
            'model_architecture': 'dense' if not hasattr(self, 'X_train_lstm') else 'lstm'
        }

        with open(components_path, 'wb') as f:
            pickle.dump(components, f)

        print(f"Model saved to Google Drive: {model_path}")
        print(f"Components saved to Google Drive: {components_path}")

        return model_path, components_path

    def load_trained_model(self, model_path, components_path):
        """
        Load complete trained model for deployment
        """
        import pickle
        from tensorflow.keras.models import load_model

        # Load model
        self.model = load_model(model_path)

        # Load components
        with open(components_path, 'rb') as f:
            components = pickle.load(f)

        self.scaler = components['scaler']
        self.target_type = components['target_type']
        self.feature_names = components['feature_names']
        self.health_thresholds = components['health_thresholds']

        print(f"Model loaded from: {model_path}")
        print(f"Components loaded from: {components_path}")
        print(f"Target type: {self.target_type}")
        print(f"Features: {len(self.feature_names)}")

        return True


def main():
    """
    Main execution function - loads preprocessed dataset and saves to new folder
    """
    import os
    from datetime import datetime

    # Path to the previously saved split dataset
    DATASET_PATH = "/content/drive/MyDrive/BearingRUL_SavedData/train_val_test_split_20250924_035736.pkl"  # Update with your actual path

    # Create new folder for this run
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    NEW_SAVE_PATH = f"/content/drive/MyDrive/fyp/Bearing_Models_{timestamp}"
    os.makedirs(NEW_SAVE_PATH, exist_ok=True)

    # Initialize the system without zip file (since we're loading preprocessed data)
    predictor = BearingRULPredictor(zip_file_path=None) # Pass None here

    try:
        # Step 1: Load preprocessed split dataset
        predictor.load_dataset(DATASET_PATH, stage='split')

        # Step 2: Build model
        predictor.build_model(model_type='lstm')  # Try 'lstm' for LSTM model

        # Step 3: Train model (will save best model to new folder)
        history = predictor.train_model(epochs=100, patience=20)

        # Save training history
        history_filepath = os.path.join(NEW_SAVE_PATH, f'training_history_{timestamp}.pkl')
        with open(history_filepath, 'wb') as f:
            pickle.dump(history.history, f)
        print(f"✅ Training history saved to: {history_filepath}")

        # Step 4: Evaluate model
        results = predictor.evaluate_model()

        # Step 5: Save final model components to new folder
        model_path, components_path = predictor.save_model_components(save_path=NEW_SAVE_PATH)

        print("\n🎉 Training completed successfully!")
        print(f"Final MAE: {results['mae']:.2f}")
        print(f"Final Accuracy: {results['accuracy']:.3f}")
        print(f"\n💾 All files saved to new Google Drive folder: {NEW_SAVE_PATH}")
        print(f"  - Best model: {model_path}")
        print(f"  - Model components: {components_path}")
        print(f"  - Training History: {history_filepath}")

    except Exception as e:
        print(f"❌ Error: {e}")
        print("Please check your file paths and data format.")

if __name__ == "__main__":
    main()
````

## Generate Windowing Illustration Diagram

### Subtask:
Create a synthetic time series to visually demonstrate the concept of applying overlapping windows. The diagram will highlight different windows and their overlap.


**Reasoning**:
I need to create a synthetic time series, define windowing parameters, and then plot the time series with highlighted overlapping windows to visually demonstrate the concept as requested by the subtask.



````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a synthetic time series
total_points = 200
t = np.arange(total_points)
synthetic_signal = np.sin(t / 10) * 5 + np.random.randn(total_points) * 0.5 + t / 20

# 2. Define parameters for windowing
window_size = 50  # number of data points in each window
overlap_percentage = 0.5 # 50% overlap

# 3. Calculate the step_size based on window size and overlap
step_size = int(window_size * (1 - overlap_percentage))

# 4. Iterate through the synthetic time series to define windows
windows = []
for i in range(0, total_points - window_size + 1, step_size):
    windows.append((i, i + window_size))
    # Only show a few windows for clarity in the diagram
    if len(windows) >= 4:
        break

# 5. Plot the original synthetic time series
plt.figure(figsize=(15, 6))
plt.plot(t, synthetic_signal, label='Synthetic Time Series', color='gray', linestyle='--')

# 6. On the same plot, highlight each window
colors = ['red', 'green', 'blue', 'purple', 'orange', 'cyan']

for idx, (start, end) in enumerate(windows):
    if idx < len(colors):
        plt.axvspan(start, end, color=colors[idx], alpha=0.2, label=f'Window {idx+1}')
        plt.plot(t[start:end], synthetic_signal[start:end], color=colors[idx], linewidth=2)
        plt.axvline(start, color=colors[idx], linestyle=':', alpha=0.7)
        plt.axvline(end, color=colors[idx], linestyle=':', alpha=0.7)

# Add markers for the start and end of each plotted window segment for better visibility
for idx, (start, end) in enumerate(windows):
    if idx < len(colors):
        plt.plot(t[start], synthetic_signal[start], 'o', color=colors[idx], markersize=8, markeredgecolor='black')
        plt.plot(t[end-1], synthetic_signal[end-1], 'x', color=colors[idx], markersize=8, markeredgecolor='black')


# 7. Add a title and labels to the plot for clarity
plt.title(f'Overlapping Windowing Illustration (Window Size: {window_size}, Overlap: {overlap_percentage*100:.0f}%)', fontsize=16)
plt.xlabel('Time Points', fontsize=12)
plt.ylabel('Signal Amplitude', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
````

## Generate Chronological vs Shuffled Split Diagram

### Subtask:
Create a diagram comparing two data splitting strategies: a standard shuffled split and a chronological split. This will illustrate how the chronological split maintains the temporal order of data, preventing data leakage in time-series datasets.


**Reasoning**:
I need to generate a synthetic time series, then create a two-panel plot illustrating a shuffled data split on one panel and a chronological data split on the other. This requires generating data, applying splitting logic conceptually for visualization, and using `matplotlib` for plotting with appropriate labels and legends.



````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a synthetic time series
total_points = 100
time_points = np.arange(total_points)
synthetic_signal = 0.1 * time_points + 5 * np.sin(time_points / 5) + np.random.randn(total_points) * 2

# 2. Define parameters for shuffled split (conceptual)
test_size_shuffled = 0.2
np.random.seed(42) # for reproducibility
shuffled_indices = np.random.permutation(total_points)
shuffled_test_indices = shuffled_indices[:int(total_points * test_size_shuffled)]
shuffled_train_indices = shuffled_indices[int(total_points * test_size_shuffled):]

# 3. Define parameters for chronological split
train_ratio_chrono = 0.7
val_ratio_chrono = 0.15
test_ratio_chrono = 0.15

train_cutoff = int(total_points * train_ratio_chrono)
val_cutoff = int(total_points * (train_ratio_chrono + val_ratio_chrono))

chrono_train_indices = time_points[:train_cutoff]
chrono_val_indices = time_points[train_cutoff:val_cutoff]
chrono_test_indices = time_points[val_cutoff:]

# 4. Plotting
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(18, 12), sharex=True)

# --- Shuffled Split Visualization ---
ax1.plot(time_points, synthetic_signal, color='gray', linestyle='--', label='Full Time Series')
ax1.scatter(time_points[shuffled_train_indices], synthetic_signal[shuffled_train_indices], color='blue', label='Train Data (Shuffled)', alpha=0.7)
ax1.scatter(time_points[shuffled_test_indices], synthetic_signal[shuffled_test_indices], color='red', label='Test Data (Shuffled)', alpha=0.7)
ax1.set_title('Standard (Shuffled) Data Split', fontsize=16)
ax1.set_ylabel('Signal Amplitude', fontsize=12)
ax1.legend()
ax1.grid(True, linestyle=':', alpha=0.6)

# --- Chronological Split Visualization ---
ax2.plot(time_points, synthetic_signal, color='gray', linestyle='--', label='Full Time Series')

ax2.axvspan(0, train_cutoff, color='lightgreen', alpha=0.4, label='Train Data (Chronological)')
ax2.axvspan(train_cutoff, val_cutoff, color='yellow', alpha=0.4, label='Validation Data (Chronological)')
ax2.axvspan(val_cutoff, total_points, color='salmon', alpha=0.4, label='Test Data (Chronological)')

ax2.axvline(train_cutoff, color='green', linestyle='--', linewidth=1.5, label='Train/Val Split')
ax2.axvline(val_cutoff, color='red', linestyle='--', linewidth=1.5, label='Val/Test Split')

ax2.text(train_cutoff / 2, ax2.get_ylim()[1] * 0.9, 'Train', ha='center', va='top', color='darkgreen', fontsize=10)
ax2.text((train_cutoff + val_cutoff) / 2, ax2.get_ylim()[1] * 0.9, 'Validation', ha='center', va='top', color='orange', fontsize=10)
ax2.text((val_cutoff + total_points) / 2, ax2.get_ylim()[1] * 0.9, 'Test', ha='center', va='top', color='darkred', fontsize=10)

ax2.set_title('Chronological Data Split', fontsize=16)
ax2.set_xlabel('Time Points', fontsize=12)
ax2.set_ylabel('Signal Amplitude', fontsize=12)
ax2.legend(loc='lower left')
ax2.grid(True, linestyle=':', alpha=0.6)

plt.suptitle('Comparison of Data Splitting Strategies for Time-Series Data', fontsize=20, y=1.02)
plt.tight_layout(rect=[0, 0, 1, 0.98]) # Adjust layout to prevent suptitle overlap
plt.show()
````

The subtask requires identifying highly correlated features with 'rul' while excluding direct proxies. I will first define the list of proxy columns to exclude. Then, I will filter the original `df_features` to get numerical features, excluding these proxies, before calculating correlations with 'rul'. Finally, I will filter by the correlation threshold and explicitly add 'rul' back.



````python
import numpy as np

# 1. Create a list of columns to exclude from the initial correlation analysis
proxy_cols = ['hour', 'health_percentage', 'health_status']

# Combine with 'rul' itself to exclude from feature selection for correlation calculation
# But 'rul' will be added back later.
all_exclude_cols = proxy_cols + ['rul']

# 2. Select numerical features from df_features, excluding proxy columns
# Filter out only numerical columns first, then exclude the specific proxy columns.
numerical_cols = df_features.select_dtypes(include=[np.number]).columns.tolist()
features_for_corr_analysis = [col for col in numerical_cols if col not in all_exclude_cols]

df_features_for_analysis = df_features[features_for_corr_analysis + ['rul']]

# 3. Calculate the absolute Pearson correlation coefficient between each selected feature and 'rul'
correlation_with_rul = df_features_for_analysis.corr(numeric_only=True)['rul'].abs()

# Define the high correlation threshold
high_corr_threshold = 0.6

# 4. Filter features with absolute correlation > 0.6 (excluding 'rul's self-correlation initially)
highly_correlated_features = correlation_with_rul[correlation_with_rul > high_corr_threshold].index.tolist()

# Remove 'rul' from the list if it accidentally got included due to self-correlation > threshold
if 'rul' in highly_correlated_features:
    highly_correlated_features.remove('rul')

# 5. Add 'rul' back to the list explicitly for the purpose of focused heatmap
highly_correlated_features.append('rul')

print(f"Features highly correlated with 'rul' (abs corr > {high_corr_threshold}), excluding proxies:")
print(highly_correlated_features)
````
